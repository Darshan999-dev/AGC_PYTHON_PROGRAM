## 1 · Why this matters

A list (notebook 12) is a *changeable* collection: you can add, remove and re-order items. But a lot of data is not like that. A point on a map is always an `(x, y)` pair. A colour is always `(red, green, blue)`. A date is always `(year, month, day)`. These are small, **fixed-size groups** of values that belong together, and you never want to append a fourth coordinate by accident.

Python has a type for exactly this: the **tuple** (say "too-pul" or "tup-ul", both are fine). It looks and behaves like a list, except that it **cannot be changed** after it is created. That one restriction makes tuples safer for fixed groupings, and it unlocks a very handy trick called **unpacking**, which you will use constantly in `for` loops and when working with functions.

## 2 · Creating and reading tuples

A tuple is written with parentheses and commas:

````python
point = (3, 7)
colour = (255, 128, 0)
person = ("Ada", 36, "London")   # items can be of different types

print(point)
print(type(point))
print(len(colour))
````

Everything you learned about reading a list works on a tuple: **indexing**, **negative indexing**, **slicing**, `len`, `in`, and looping. (Slicing a tuple gives a new tuple.)

````python
print(person[0])        # first item
print(person[-1])       # last item
print(colour[:2])       # slice: a new tuple
print(128 in colour)    # membership

for channel in colour:
    print(channel)
````

Two special cases trip people up:

- The **empty tuple** is `()`.
- A **one-item tuple** needs a trailing comma: `(5,)`. It is the *comma* that makes a tuple, not the parentheses. Without the comma, `(5)` is just the number 5 in brackets.

You can also turn another sequence into a tuple with `tuple(...)`:

````python
empty = ()
single = (5,)
not_a_tuple = (5)

print(type(empty), type(single), type(not_a_tuple))

print(tuple([1, 2, 3]))      # from a list
print(tuple("abc"))          # from a string
print(tuple(range(4)))       # from a range
````

Tuples have just two methods, both for reading: `count` (how many times does a value appear) and `index` (where does it first appear).

````python
marks = (7, 3, 7, 9, 7)
print(marks.count(7))
print(marks.index(9))
````

## 3 · Immutability

**Immutable** means "cannot be changed after creation". Tuples (like strings and numbers) are immutable; lists are **mutable**. You can *read* a tuple in every way you can read a list, but you cannot assign to an item, append, remove or sort it in place. ⚠️ **Expected to fail.**

````python
point = (3, 7)
point[0] = 10
````

Python refuses with a `TypeError`. If you want a "changed" tuple, you build a **new** one, for example by slicing and joining with `+`:

````python
point = (3, 7)
moved = (10,) + point[1:]    # new tuple: replace the first item
print(point)    # unchanged
print(moved)

longer = point + (9,)        # join two tuples
print(longer)
````

Notice that `point` itself never changed: tuples can only be *replaced* (by rebinding the name to a new tuple), never edited in place.

This also removes the aliasing worry from notebook 12. If two names refer to the same tuple, nobody can change it behind your back:

````python
a = (1, 2, 3)
b = a
b = b + (4,)     # makes a new tuple and rebinds b
print(a)
print(b)
````

## 4 · When to prefer a tuple

Use a **list** for a collection that grows, shrinks or is re-ordered, and where all items play the same role (a list of scores, a list of names).

Use a **tuple** for a **fixed-size record**, where each *position* has its own meaning: `(x, y)`, `(red, green, blue)`, `(name, age, city)`. Reasons:

- It tells the reader (and Python) that the structure is not meant to change.
- It prevents accidents such as appending a third coordinate to a point.
- Because it can't change, tuples can be used in places where Python needs a value that stays fixed, for instance as keys in a dictionary (notebook 14).
- A tuple is a little smaller and faster than the equivalent list. Not a reason to choose it, but a pleasant bonus.

A useful rule of thumb: *if you would write it as a row in a table with named columns, it's a tuple; if you would write it as a column, it's a list.* A list of tuples is a very common shape:

````python
students = [("Ada", 36), ("Grace", 85), ("Alan", 41)]   # list of (name, age) tuples
students.append(("Linus", 55))                          # the list can grow
print(students)
print(students[1])
print(students[1][0])
````

### A catch: immutable does not mean "deeply frozen"

A tuple always holds the *same items*, but if one of those items is itself mutable (such as a list), that item can still change:

````python
record = ("Ada", [90, 85])
record[1].append(100)      # changes the list inside; the tuple still holds the same list
print(record)
````

No error here: we never tried to replace an item of the tuple, we changed the list that the item refers to. It's rare in practice, but good to know.

## 5 · Packing and unpacking

Writing several values separated by commas **packs** them into a tuple. The parentheses are optional:

````python
packed = 3, 7
print(packed, type(packed))
````

The reverse is **unpacking**: assign a tuple (or any sequence) to several variables at once, and each variable receives the item in the matching position.

````python
point = (3, 7)
x, y = point
print(x)
print(y)

name, age, city = ("Ada", 36, "London")
print(f"{name} is {age} and lives in {city}.")
````

Unpacking is more than convenience: `x, y = point` is far more readable than `point[0]` and `point[1]`, because the variable names say what each item means.

### Swapping two variables

The right-hand side is fully packed *before* the left-hand side is assigned, which gives the classic one-line swap, with no temporary variable:

````python
a = 1
b = 2
a, b = b, a
print(a, b)
````

### Unpacking in `for` loops

You have already seen this in notebook 10: `for i, letter in enumerate(...)` and `for a, b in zip(...)` unpack a pair on every repetition. It works for any sequence of tuples:

````python
students = [("Ada", 36), ("Grace", 85), ("Alan", 41)]

for name, age in students:
    print(f"{name} is {age}")
````

### Ignoring items with `_`

By convention, the name `_` means "I don't care about this one":

````python
name, _, city = ("Ada", 36, "London")
print(name, city)
````

### The counts must match

If the number of variables differs from the number of items, Python raises a `ValueError` (see the common mistakes below). Unpacking works on lists and strings too, not only tuples, as long as the counts match.

````python
first, second, third = [10, 20, 30]
c1, c2, c3 = "xyz"
print(first, second, third, c1, c2, c3)
````

## 6 · `*` unpacking: head, middle and tail

What if you don't know (or don't care about) how many items there are? Put a `*` in front of **one** variable, and it collects "all the rest" into a **list**:

````python
numbers = (10, 20, 30, 40, 50)

head, *tail = numbers
print(head)
print(tail)
````

````python
*start, last = numbers
print(start)
print(last)

first, *middle, last = numbers
print(first, middle, last)
````

Things to notice:

- The starred variable always receives a **list** (even if the source was a tuple), and it may be empty.
- Only **one** starred variable is allowed in a single unpacking.
- It works on any sequence, including strings.

````python
only, *rest = (99,)
print(only, rest)          # rest is empty

letter, *others = "python"
print(letter, others)
````

The `*` also works the other way round, to **spread** a sequence into a call or into a new collection:

````python
parts = ("2026", "10", "19")
print(*parts)                  # same as print("2026", "10", "19")
print(*parts, sep="-")

combined = (*parts, "extra")   # build a new tuple from the pieces
print(combined)
````

## 7 · Returning multiple values (a preview)

Sometimes one operation naturally produces *two* answers. Python's way to hand back several values is to return them **packed in a tuple**, and the caller unpacks them. You can already see this with a built-in: `divmod(a, b)` returns the quotient and the remainder together.

````python
result = divmod(17, 5)
print(result, type(result))

quotient, remainder = divmod(17, 5)
print(quotient, remainder)
````

The string method `partition` is another: it splits text at the first occurrence of a separator and returns a 3-tuple `(before, separator, after)`.

````python
user, at, domain = "ada@example.com".partition("@")
print(user)
print(domain)
````

## 8 · Common mistakes

**Mistake 1: forgetting the comma in a one-item tuple.**

````python
single = (42)
print(type(single))       # int, not tuple!
fixed = (42,)
print(type(fixed))
````

**Mistake 2: trying to change an item of a tuple.** The error is a `TypeError`. ⚠️ **Expected to fail.**

````python
colour = (255, 128, 0)
colour[2] = 255
````

Same for methods that modify in place, such as `append`: tuples don't have them. To get a longer tuple, join with `+` to make a new one. ⚠️ **Expected to fail.**

````python
colour = (255, 128, 0)
colour.append(10)
````

**Mistake 3: unpacking into the wrong number of variables.** Python tells you how many it expected. ⚠️ **Expected to fail.**

````python
x, y = (1, 2, 3)
````

⚠️ **Expected to fail.**

````python
x, y, z = (1, 2)
````

The fix is to match the counts, or to use `*` to collect the extras (`x, y, *rest = (1, 2, 3)`), or `_` to ignore one.

**Mistake 4: expecting a starred variable to be a tuple.** It is always a list.

````python
first, *rest = (1, 2, 3)
print(rest, type(rest))
````

**Mistake 5: assuming `+=` edits a tuple.** It looks like it works, but it creates a *new* tuple and rebinds the name; the old one is untouched (see `a` and `b` earlier). That is fine, but it means that building a big tuple one item at a time is clumsy. If you need to keep adding things, use a list.

Run it and compare `t` with `same`:

````python
t = (1, 2, 3)
same = t
t += (4,)          # builds a NEW tuple and rebinds t
print(t)           # (1, 2, 3, 4)
print(same)        # (1, 2, 3): the original tuple is unchanged
````

**Mistake 6: swapping the "wrong" way.** Writing `a = b` then `b = a` loses the original `a`. Use `a, b = b, a`.

````python
a, b = 1, 2
a = b
b = a
print(a, b)     # both are 2: the original 1 is gone
````

## 10 · Recap and what's next

- A **tuple** is an ordered, **immutable** sequence: `(3, 7)`. You can index, slice, loop, `len`, `in`, `count` and `index`, but not assign to items or append.
- A one-item tuple needs a trailing comma, `(5,)`; the empty tuple is `()`. `tuple(x)` converts another sequence.
- "Changing" a tuple means building a **new** one (slicing and `+`). If an item inside is mutable, that item can still change.
- Prefer a tuple for a **fixed-size record** where position has meaning; prefer a list for a collection that grows or changes.
- **Packing** (`t = 1, 2`) and **unpacking** (`x, y = t`) are everywhere: swapping (`a, b = b, a`), `for name, age in pairs`, and ignoring items with `_`. Counts must match, or you get a `ValueError`.
- **`*`** collects the rest into a list (`first, *rest = seq`, `first, *middle, last = seq`) and spreads a sequence into a call (`print(*parts)`).
- Returning several values is returning a tuple that the caller unpacks (`divmod`, and your own functions in notebook 17).
