## 1 · Why this matters

A program that does the same thing no matter what is a very fancy `print`. Decisions are what make software respond: a login screen that behaves differently for a right and a wrong password, a game that reacts to your move, a shop that applies a discount only above a certain total.

A **conditional** (or `if` statement) runs a block of code only when a condition is true. The condition is any boolean question from notebook 07.

## 2 · `if`

```
if condition:
    code that runs when the condition is true
```

Three things to notice:

1. The condition is followed by a **colon** `:`.
2. The code to run is **indented** (by convention 4 spaces). The indentation *is* the syntax; it is how Python knows which lines belong to the `if`.
3. When the indented block ends (the next line is back at the left margin), the `if` is over and the rest of the program continues as normal.

````python
temperature = 31

if temperature > 30:
    print("It's hot!")
    print("Drink some water.")

print("Have a nice day.")      # not indented: runs either way
````

Change `temperature` to `20` and run it again: the two indented lines are skipped, but "Have a nice day" always prints.

> 💡 **In Jupyter**, pressing Enter after a colon indents the next line automatically, and `Tab` / `Shift+Tab` indent / unindent selected lines.

## 3 · `else`: the other case

`else` runs when the condition was false. Exactly one of the two blocks runs.

````python
password = "letmein"

if password == "swordfish":
    print("Access granted")
else:
    print("Access denied")
````

## 4 · `elif`: more than two outcomes

When there are several mutually exclusive cases, chain them with `elif` ("else if"). Python tests the conditions **from the top down and runs the first branch that is true**, then skips all the rest. An optional final `else` catches everything left.

````python
score = 83

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print("Grade:", grade)
````

Try `score` values of 95, 80, 70, 69 and 0.

### The order matters

Because the first true branch wins, put the **most specific** test first. Here's the same grading in the wrong order:

````python
score = 95

if score >= 70:
    grade = "C"
elif score >= 80:
    grade = "B"
elif score >= 90:
    grade = "A"
else:
    grade = "F"

print("Grade:", grade)       # a 95 gets a C!
````

`score >= 70` is true for 95, so Python never looks at the other branches. A good habit: after writing an `elif` chain, test it with a value on **every boundary** (here 90, 89, 80, 79, 70, 69).

### `elif` versus several `if`s

A chain of `elif`s runs at most one block. Separate `if` statements are independent: each one runs if its own condition is true.

````python
x = 15

# chain: only the first match runs
if x > 10:
    print("big")
elif x > 5:
    print("medium")

# separate ifs: both can run
if x > 10:
    print("big")
if x > 5:
    print("medium")
````

## 5 · Indentation in detail

- A block is every line indented *the same amount* after the colon.
- Use **4 spaces** per level. Don't mix tabs and spaces.
- Blocks can contain more `if`s, which are indented further (nested).
- Python reports `IndentationError` if the indentation doesn't make sense (see Common mistakes).

### Nested conditionals

You can place an `if` inside another:

````python
age = 20
has_ticket = True

if age >= 18:
    if has_ticket:
        print("Welcome in!")
    else:
        print("You need a ticket.")
else:
    print("Sorry, adults only.")
````

### Nested versus combined

The same logic can often be written with `and`:

````python
age = 20
has_ticket = True

if age >= 18 and has_ticket:
    print("Welcome in!")
else:
    print("Cannot enter.")
````

Which is better? Prefer the **flat, combined** version when you only have one outcome for "everything else". Use **nesting** when different failures need different messages (as in the first version), but avoid going more than two or three levels deep: deeply nested code is hard to read.

A common technique to flatten code is to deal with the unusual cases first:

````python
age = 15
has_ticket = True

if age < 18:
    print("Sorry, adults only.")
elif not has_ticket:
    print("You need a ticket.")
else:
    print("Welcome in!")
````

## 6 · Conditions can be any expression

## 7 · The conditional expression

When you only need to choose between two **values**, there's a compact form:

```
value_if_true if condition else value_if_false
```

````python
score = 72
label = "pass" if score >= 50 else "fail"
print(label)

age = 17
print("adult" if age >= 18 else "minor")
````

It's an *expression*, so it can appear anywhere a value can, such as inside an f-string or a function call. It reads like English, but don't chain several together; use a regular `if`/`elif` for that.

## 8 · `match` (optional)

Python 3.10 added `match` for comparing one value against several **patterns**. It's the tidy way to write a long chain of `==` comparisons:

````python
command = "stop"

match command:
    case "go":
        print("Moving")
    case "stop":
        print("Stopping")
    case "left" | "right":          # either of two values
        print("Turning")
    case _:                         # underscore = anything else (like else)
        print("Unknown command")
````

You can also add a condition (a *guard*) to a case:

````python
number = -4

match number:
    case 0:
        print("zero")
    case n if n < 0:
        print("negative:", n)
    case _:
        print("positive")
````

`match` can do much more (matching the shape of lists and objects) but a long `if`/`elif` chain does the same job for what we need now. Everything in this notebook except this section is essential; this one is a convenience.

### 6. A condition after `else`

`else` never takes a condition; use `elif` for that. ⚠️ **Expected to fail.**

````python
x = 1
if x == 1:
    print("one")
else x == 2:
    print("two")
````

### 7. Mistakes that don't crash

These run but give wrong answers, which is worse.

- **Wrong order** in an `elif` chain (see section 4).
- **`if` where you wanted `elif`**, so several blocks run.
- **`if x == 1 or 2`** is always true (notebook 07). Write `x == 1 or x == 2`.
- **Boundary errors**: `>` where `>=` was meant. Test the exact boundary value.
- **Comparing text to numbers**: `if answer == 56` where `answer` came from `input()` is never true, because `"56" != 56`. Convert first.

## 11 · Recap and what's next

- `if condition:` runs an **indented block** only when the condition is true. Colons and indentation are part of the syntax.
- `else` handles the other case; `elif` adds more cases. The **first true branch wins**, so order matters. Test every boundary.
- Separate `if`s are independent; an `elif` chain runs at most one branch.
- Combine conditions with `and`/`or` for flat code; nest when different failures need different responses, but keep nesting shallow.
- `a if condition else b` picks between two values; `match` tidies up long `==` chains.
