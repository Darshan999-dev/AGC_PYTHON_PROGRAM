## 1 · Why this matters

Computers are good at repeating things. Without loops you would write the same line a hundred times to count to a hundred, and you couldn't handle "keep asking until the user types something valid" at all, because you don't know in advance how many tries it will take.

A **`while` loop** repeats a block of code *as long as a condition stays true*. It is the right tool when you don't know in advance how many repetitions you need.

## 2 · Anatomy of a `while` loop

```
while condition:
    body (indented)
```

Python checks the condition. If it's true, it runs the body, then goes **back to the top and checks again**. When the condition is false, the loop ends and the program continues after it.

````python
count = 1                    # 1. initialise: set up before the loop

while count <= 5:            # 2. condition: keep going while this is true
    print("count is", count) # 3. body: the work
    count += 1               # 4. update: move towards finishing

print("Done! count ended at", count)
````

Every well-behaved loop has all four parts:

1. **Initialise** the variable(s) the condition depends on.
2. **Condition** that is true while there is more to do.
3. **Body**, the work for one repetition.
4. **Update** something so the condition eventually becomes false.

### Tracing a loop

Trace it like in notebook 02, one repetition at a time:

| Check | `count` | `count <= 5`? | Action |
|---|---|---|---|
| 1st | 1 | True | print 1, count becomes 2 |
| 2nd | 2 | True | print 2, count becomes 3 |
| ... | | | |
| 5th | 5 | True | print 5, count becomes 6 |
| 6th | 6 | **False** | loop ends |

Notice that the loop ends with `count` at 6, one past the last value printed. That "one past" is a common source of off-by-one errors.

### A loop that may not run at all

If the condition is false the first time, the body is skipped completely:

````python
count = 10
while count <= 5:
    print("this never prints")
print("Skipped the loop entirely")
````

## 3 · Accumulators: keeping a running total

The most common thing to do inside a loop is to **accumulate** a result: start with a neutral value (`0` for a sum, `1` for a product, `""` for text) and update it each time round.

````python
total = 0
n = 1

while n <= 10:
    total += n
    n += 1

print("Sum of 1 to 10:", total)
````

````python
# Doubling: how many days until savings pass 1000, starting at 1 and doubling each day?
amount = 1
days = 0

while amount < 1000:
    amount *= 2
    days += 1

print(days, "days:", amount)
````

Counting how many times something happens is the same idea (`days += 1`). Here the loop runs "until a goal is reached", which is the kind of problem `while` is made for, since we couldn't easily have predicted `10` in advance.

## 4 · Infinite loops

What if the condition never becomes false? The loop runs forever. This is the single most common `while` bug, and nearly everybody writes one in their first week. It's not a disaster: it just means the cell keeps spinning.

**How to stop a runaway loop in Jupyter:** click the **■ Interrupt** button in the toolbar (or press `I` twice in command mode, or use **Kernel → Interrupt Kernel**). If that fails, **Kernel → Restart**.

Here's a classic bug: we forgot the *update*. To keep this demo safe we've added a `guard` counter that forces the loop to stop after five goes. The condition `n <= 3` itself would never become false:

````python
n = 1
guard = 0

while n <= 3 and guard < 5:
    print("n is still", n)     # we forgot n += 1 !
    guard += 1

print("The guard stopped us. Without it, this would never end.")
````

### Loops that are intentionally infinite

Sometimes you *want* an endless loop, such as a game's main loop or a program waiting for the user to quit. The idiom is `while True:` plus a `break` that fires on the exit condition:

````python
n = 1
while True:
    print(n, "squared is", n * n)
    if n * n > 50:
        break                    # leave the loop immediately
    n += 1

print("Stopped at n =", n)
````

Here the condition is simply `True`, so only the `break` can end the loop. Make sure there *is* one that will fire!

## 5 · `break` and `continue`

- **`break`** ends the loop immediately.
- **`continue`** skips the rest of *this repetition* and jumps back to the condition check.

````python
n = 0
while n < 10:
    n += 1
    if n % 2 == 0:
        continue          # skip even numbers
    if n > 7:
        break             # stop altogether
    print(n)
````

Reading it: it prints the odd numbers 1, 3, 5, 7. Then `n` becomes 9, which is odd but greater than 7, so we `break`.

> ⚠️ **Where you put the update matters with `continue`.** If `continue` jumps back *before* the update runs, `n` never changes and you get an infinite loop. In the example above, we moved the update (`n += 1`) to the top of the body to avoid exactly that.

### `while ... else`

A `while` loop can have an `else` block that runs when the condition becomes false **without a `break`**. It runs if the loop finished "naturally" and is skipped if the loop was exited early. It reads oddly, but it is perfect for "search, and tell me if you didn't find it":

````python
# Is there a multiple of 7 between 90 and 96?
n = 90
while n <= 96:
    if n % 7 == 0:
        print("Found one:", n)
        break
    n += 1
else:
    print("No multiple of 7 in that range")
````

Change the range to `90..90` and re-run to see the `else` branch fire.

## 6 · Sentinel loops

A **sentinel** is a special value that means "stop". The loop keeps asking until the user types it.

````python
total = 0

text = input("Enter a number (or 'done'): ")
while text != "done":
    total += int(text)
    text = input("Enter a number (or 'done'): ")

print("Total:", total)
````

Notice the pattern: **ask once before the loop**, then **ask again at the end of the body**. That way the condition always has a fresh answer to test. (An alternative is `while True` with an `if ...: break`.)

## 7 · Input validation

A `while` loop is the standard way to say "keep asking until the answer is acceptable". Strings have a handy method, `.isdigit()`, that is `True` if the text consists only of digits, so we can check *before* converting and avoid a `ValueError`:

````python
print("42".isdigit(), "4.2".isdigit(), "-5".isdigit(), "abc".isdigit(), "".isdigit())
````

````python
text = input("Enter your age: ")
while not text.isdigit():
    print("That's not a whole number. Try again.")
    text = input("Enter your age: ")

age = int(text)
print("Age accepted:", age)
````

(`.isdigit()` doesn't accept a minus sign or decimal point, so this only accepts non-negative whole numbers. Notebook 21 shows a more flexible approach using exceptions.)

## 8 · A menu loop

Many text programs show a menu, do what the user asks, and repeat until they quit:

````python
counter = 0

while True:
    choice = input("(a)dd, (s)how, (q)uit: ")
    if choice == "a":
        counter += 1
        print("Added. Counter is", counter)
    elif choice == "s":
        print("Counter is", counter)
    elif choice == "q":
        print("Bye!")
        break
    else:
        print("Unknown command")
````

## 9 · Common mistakes

### 1. Forgetting the update (infinite loop)

Seen above. Always ask: *what changes each time round, and does it move the condition towards false?*

### 2. Update in the wrong place

If the update is **indented too little**, it's outside the loop and never runs inside it. If the update comes after a `continue`, it gets skipped.

### 3. Off-by-one: `<` versus `<=`

`while n < 5` runs for n = 0..4; `while n <= 5` runs for n = 0..5. Trace the first and last repetitions to check.

````python
n = 1
while n < 5:
    print(n, end=" ")
    n += 1
print()

n = 1
while n <= 5:
    print(n, end=" ")
    n += 1
print()
````

### 4. Forgetting to initialise

⚠️ **Expected to fail.**

````python
while never_defined_counter < 3:
    print("hello")
    never_defined_counter += 1
````

### 5. A condition that is false from the start

If you write `while count == 5:` when `count` starts at 0, the loop never runs. If your loop "does nothing", print the condition's value before the loop.

### 6. Comparing floats for exact equality

Floating-point rounding means `x != 1.0` may never become false when counting by 0.1. Use `<` or `<=`, or count with integers.

````python
x = 0.0
steps = 0
while x < 1.0:        # '<' is safe; 'x != 1.0' might never be reached
    x += 0.1
    steps += 1
print(steps, x)
````

Notice that it took 11 steps, not 10, and `x` ended up as `1.0999999999999999`: the accumulated rounding error from notebook 03 in action.

### 7. `=` instead of `==`

`while x = 5:` is a `SyntaxError`. You want `==`, or more often a `<`/`>` comparison.

### 8. Using `while` where `if` is enough

An `if` runs its body once; a `while` keeps going. If you only wanted a single check, use `if`.

## 11 · Recap and what's next

- `while condition:` repeats its body **as long as the condition is true**. Four parts: initialise, condition, body, update.
- Forgetting the update gives an **infinite loop**: interrupt the kernel to stop it.
- `break` leaves the loop; `continue` skips to the next check; `else` runs if the loop ended without a `break`.
- Patterns: **accumulators** (running totals, counters), **sentinel loops** (ask until `"done"`), **validation loops** (ask until acceptable) and **menu loops** (`while True` + `break`).
- Always trace the first and last repetitions and watch `<` versus `<=`.
