## 1 · Why this matters

Imagine you need to turn a temperature from Celsius to Fahrenheit in five places in your program. You *could* copy and paste the formula five times. But then, when you discover a mistake in it, you must find and fix all five copies, and you will miss one.

A **function** is a named, reusable piece of code. You write it once, give it a name, and then use that name whenever you need it. Functions let you:

- **Reuse** code instead of copying it.
- **Name** an idea (`celsius_to_fahrenheit` says more than `c * 9 / 5 + 32`).
- **Test** a small piece on its own, before putting it into a bigger program.
- **Break big problems into small ones**, so no single piece is hard to understand.

## 2 · Defining and calling a function

You **define** a function with the keyword `def`, a name, parentheses, and a colon. The indented lines below it are the **body**: the code that runs each time the function is used.

````python
def say_hello():
    print("Hello!")
    print("Welcome to functions.")
````

Running that cell printed nothing. Defining a function only *teaches Python* what the name means; it doesn't run the body. To run it, **call** the function by writing its name followed by parentheses:

````python
say_hello()
say_hello()
````

Each call runs the body once from the top. The parentheses are what make it a call. The anatomy:

```
def say_hello():          # keyword, name, parentheses, colon
    print("Hello!")       # body: indented, runs on each call
```

Function names follow the same rules as variable names: lowercase words joined by underscores (`say_hello`, `celsius_to_fahrenheit`). Pick a name that says what the function *does*.

### Information in: parameters and arguments

A function that always does exactly the same thing isn't very useful. We make it flexible with **parameters**: names listed inside the parentheses of the `def`, which act like variables that get their values each time the function is called.

The actual values you give when calling are the **arguments**.

````python
def greet(name):              # name is a PARAMETER
    print("Hello,", name + "!")

greet("Ada")                  # "Ada" is an ARGUMENT
greet("Grace")
````

Remember it this way: the **parameter** is the *name* in the `def`; the **argument** is the *value* you pass in the call. On the first call, Python does the equivalent of `name = "Ada"` and then runs the body.

A function can have several parameters, separated by commas. The arguments are matched **by position**: the first argument goes to the first parameter, the second to the second, and so on.

````python
def introduce(name, age):
    print(name, "is", age, "years old.")

introduce("Ada", 36)
introduce(36, "Ada")      # same code, arguments swapped: nonsense output
````

Order matters, and the number of arguments must match the number of parameters. (Python has other ways to pass arguments, such as by name or with defaults; `18_Function_Arguments_and_Scope` covers those. Here we always pass them by position.)

## 3 · `return` vs. `print`

This is the most important idea in this notebook, and the one beginners trip over most.

- `print` **shows** something on the screen. A human can read it, but your program can't use it afterwards.
- `return` **hands a value back** to the code that called the function. The program can store it, add it to something, pass it on.

Compare:

````python
def square_print(n):
    print(n * n)

def square(n):
    return n * n

square_print(4)          # shows 16 on screen
square(4)                # computes 16 and hands it back (the notebook displays it because it is the last line)
````

The notebook displayed `16` for `square(4)` because a bare call on the *last line of a cell* shows its value; in a program that value would be silently thrown away. The real difference shows when you try to *use* the answer:

````python
a = square(4)
b = square(5)
print(a + b)             # 16 + 25

c = square_print(4)      # prints 16, but what is stored in c?
print(c)
````

`c` holds `None`: `square_print` printed, but gave nothing back. (More on `None` in section 7.)

> **Rule of thumb.** If a function *computes* something, **return** it and let the caller decide whether to print it. A function that only prints is hard to test and hard to reuse.

Two more facts about `return`:

1. A function **stops** the moment it hits `return`. Any code after it in that call is skipped.
2. A `return` can send back any value: numbers, strings, lists, `True`/`False`, and so on.

````python
def describe_number(n):
    if n < 0:
        return "negative"
    if n == 0:
        return "zero"
    return "positive"          # reached only if neither earlier return ran

print(describe_number(-5), describe_number(0), describe_number(8))
````

Because the answer is *returned*, we can use the call anywhere a value is allowed: inside another expression, as an argument to another function, in an `if`:

````python
def is_even(n):
    return n % 2 == 0

if is_even(10):
    print("10 is even")
print(is_even(7), is_even(7) or is_even(8))
print(square(square(3)))       # a call inside a call: 3*3 = 9, then 9*9 = 81
````

## 4 · Local variables: names inside a function

Variables you create inside a function (including the parameters) exist only while the function is running. They are **local** to that call. This is a good thing: the function's inner workings can't accidentally clash with names elsewhere in your program.

````python
def total_with_tax(price, rate):
    tax = price * rate
    total = price + tax
    return total

print(total_with_tax(100, 0.2))
````

Here `tax` and `total` are scratch space. Once the function returns, they are gone (we will see the error you get if you try to use them outside, in the *Common mistakes* section). The *returned value* is how information gets out. (A fuller picture of where names are visible, called **scope**, is in `18_Function_Arguments_and_Scope`.)

A function body can contain anything you've learned so far: `if`, loops, lists, dicts. Here a list goes in as an argument:

````python
def total(numbers):
    result = 0
    for n in numbers:
        result += n
    return result

print(total([4, 8, 15, 16, 23, 42]))
print(total([]))               # an empty list: the loop never runs, result stays 0
````

### Returning early from a loop

Because `return` leaves the function immediately, it is a neat way to *stop searching* as soon as you find what you want. (In Level 1 we noted that `break` only leaves the innermost loop; `return` leaves everything.)

````python
def first_negative(numbers):
    for n in numbers:
        if n < 0:
            return n           # found it: leave the whole function now
    return None                # the loop finished without finding one

print(first_negative([3, 8, -2, 5, -9]))
print(first_negative([3, 8, 5]))
````

## 5 · Docstrings: explaining what a function does

A **docstring** is a string placed as the very first line of the body. It describes what the function does, so that a reader (including you, next month) doesn't have to decode the code. Use triple quotes.

````python
def celsius_to_fahrenheit(celsius):
    """Convert a temperature from degrees Celsius to degrees Fahrenheit."""
    return celsius * 9 / 5 + 32

print(celsius_to_fahrenheit(100))
````

Python stores the docstring with the function, and tools use it. The built-in `help()` shows it, and so do notebooks if you press Shift+Tab inside the parentheses of a call:

````python
help(celsius_to_fahrenheit)
print(celsius_to_fahrenheit.__doc__)
````

A good docstring says **what** the function does, what it takes and what it returns, in a sentence or two. For longer ones, put the summary on the first line and details below:

````python
def average(numbers):
    """Return the mean of a list of numbers.

    The list must not be empty.
    """
    return sum(numbers) / len(numbers)

print(average([2, 4, 9]))
````

Comments (`#`) explain *how* a tricky line works; a docstring explains *what the whole function is for*. Use both sparingly, and only where they add something the code doesn't already say.

## 6 · Why functions? DRY, naming and testing

### Don't repeat yourself (DRY)

Programmers have a motto: **Don't Repeat Yourself**. Look at this code; the same idea appears three times:

````python
print("Total for Alice:", 3 * 4.50 + 2 * 1.25)
print("Total for Bob:", 1 * 4.50 + 6 * 1.25)
print("Total for Cara:", 5 * 4.50 + 0 * 1.25)
````

If the coffee price changes, we must edit three places. Move the idea into a function:

````python
COFFEE = 4.50
BISCUIT = 1.25

def order_total(coffees, biscuits):
    """Return the cost of an order of coffees and biscuits."""
    return coffees * COFFEE + biscuits * BISCUIT

print("Total for Alice:", order_total(3, 2))
print("Total for Bob:", order_total(1, 6))
print("Total for Cara:", order_total(5, 0))
````

Now the pricing rule lives in one place. (The function also reads `COFFEE` and `BISCUIT`, which live outside it; reading such constants is fine, and `18_Function_Arguments_and_Scope` explains the rules.)

### Names make code readable

Compare `c * 9 / 5 + 32` with `celsius_to_fahrenheit(c)`. A good function name lets a reader skip the details until they need them. Choose verbs for actions (`count_vowels`, `print_report`) and questions for `True`/`False` functions (`is_even`, `has_duplicates`).

### Testing in isolation

A small function can be tested on its own with a few inputs whose answers you know. That is exactly what `check()` does in the exercises:

````python
check(order_total(3, 2), 16.0)
check(order_total(0, 0), 0.0)
check(is_even(10), True)
````

Professional programmers do the same: they write small functions, try each on easy cases, then combine them. When a combined program misbehaves, you already know the small parts work, so the bug is in how they are joined. (Level 3 has a whole notebook on testing.)

### Functions can call other functions

Build bigger tools from smaller ones:

````python
def is_adult(age):
    return age >= 18

def ticket_price(age):
    """Return the ticket price: 5 for under 18, otherwise 12."""
    if is_adult(age):
        return 12
    return 5

print(ticket_price(10), ticket_price(40))
````

## 7 · Functions that return `None`

If a function never reaches a `return` with a value, Python quietly returns the special value `None`, which means "nothing / no value". That includes functions that only `print`.

````python
def log(message):
    print("LOG:", message)

result = log("starting")
print(result)
print(type(result))
````

There are two broad kinds of functions:

| Kind | What it does | Example |
|---|---|---|
| **Computes a value** | `return`s an answer | `square(4)`, `is_even(7)` |
| **Does an action** (a "side effect") | prints, changes a list, ... and returns `None` | `log("hi")`, `list.append` |

Both are legitimate. Python's own `list.append` and `list.sort` are of the second kind, and return `None`; this is why `x = my_list.sort()` leaves `x` as `None`.

A bare `return` (with no value) leaves the function early and also gives back `None`:

````python
def print_positive(n):
    if n <= 0:
        return                # nothing to do: stop here (returns None)
    print("Positive:", n)

print_positive(-3)
print_positive(7)
````

Functions that sometimes return a value and sometimes `None` (like `first_negative` above) are common for "found it / didn't find it" situations. Just remember to handle the `None` in the code that calls them:

````python
found = first_negative([1, 2, 3])
if found is None:
    print("No negative numbers.")
else:
    print("First negative:", found)
````

## 8 · Returning multiple values

Sometimes one answer isn't enough. A function might want to give back both the smallest *and* the largest item. In `13_Tuples_and_Unpacking` you saw a preview: separate the values with commas after `return`, and Python packs them into a **tuple**.

````python
def min_and_max(numbers):
    return min(numbers), max(numbers)

result = min_and_max([5, 2, 9, 1, 7])
print(result)
print(type(result))
````

The caller can take the tuple apart with **unpacking**: put as many names on the left as there are values in the tuple.

````python
low, high = min_and_max([5, 2, 9, 1, 7])
print("low:", low)
print("high:", high)

# or index the tuple
print(min_and_max([5, 2, 9, 1, 7])[1])
````

It is really still *one* return value (a tuple); unpacking is just a convenient way to receive it. The number of names must match the number of values, otherwise you get a `ValueError`.

Here is a realistic example: splitting a total number of seconds into minutes and seconds, using `//` and `%` from Level 1:

````python
def split_time(total_seconds):
    """Return (minutes, seconds) for a number of seconds."""
    minutes = total_seconds // 60
    seconds = total_seconds % 60
    return minutes, seconds

m, s = split_time(135)
print(f"{m} min {s} s")
````

## 9 · Common mistakes

### 1. Forgetting to call: no parentheses

Writing just the name *refers* to the function but doesn't run it:

````python
say_hello          # no parentheses: Python just shows that it's a function
````

### 2. Forgetting `return` (printing instead)

The function prints the right number, but the caller gets `None` and can't calculate with it:

````python
def add_print(a, b):
    print(a + b)

total = add_print(2, 3)
print(total)
````

⚠️ **Expected to fail.**

````python
print(add_print(2, 3) * 10)     # None * 10 is an error
````

Fix: `return a + b`. When a result is wrong or `None` and you can't see why, check whether the function ends in `return`.

### 3. Wrong number of arguments

The count of arguments must match the parameters: ⚠️ **Expected to fail.**

````python
introduce("Ada")                # introduce needs two arguments
````

Read the message: it names the function and says which argument is missing. The same error appears when you pass too many.

### 4. Using a local variable outside the function

`tax` was created inside `total_with_tax`. It disappears when the function finishes: ⚠️ **Expected to fail.**

````python
total_with_tax(100, 0.2)
print(tax)
````

The fix is to `return` what you need, and store the result in a variable outside.

### 5. Code after `return` never runs

````python
def silly(n):
    return n + 1
    print("This line is never reached")

print(silly(1))
````

### 6. Returning too early inside a loop

A `return` in a loop body ends the *function* on the first pass. This version is meant to add up all the numbers:

````python
def total_buggy(numbers):
    result = 0
    for n in numbers:
        result += n
        return result       # bug: indented inside the loop

print(total_buggy([1, 2, 3]))   # should be 6
````

Fix: put `return result` *after* the loop, lined up with the `for` (as in the earlier `total`).

### 7. A forgotten colon or indentation

`def f()` without the colon, or a body that isn't indented, is a `SyntaxError` / `IndentationError`: Python refuses to run the cell at all. Check the end of the `def` line and the indentation of the body.

## 11 · Recap and what's next

- A function is defined with `def name(parameters):` and an indented body. Defining it doesn't run it; **calling** it with parentheses does.
- **Parameters** are the names in the `def`; **arguments** are the values you pass. For now arguments are matched by position.
- **`return`** hands a value back and ends the function. **`print`** only shows something. If a function computes, return it.
- A function with no `return` value gives back **`None`**. Variables created inside a function are **local** to it.
- **Docstrings** (triple-quoted first line) say what a function does; `help(f)` shows them.
- Functions let you avoid repetition (DRY), name your ideas, and **test small pieces** with `check(func(args), expected)`.
- `return a, b` returns a tuple; **unpack** it with `x, y = func(...)`.
