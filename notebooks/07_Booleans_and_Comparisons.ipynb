## 1 · Why this matters

So far a program has done every line, in order, every time. Real programs need to **decide**: *if* the password is wrong, show an error; *while* there are guesses left, keep going; *for* each customer, check their balance.

Every decision boils down to a **yes/no question** whose answer is a value Python can work with. That value is a **boolean**, and the questions are built from comparisons and logic. This notebook is the vocabulary; the next three (conditionals and loops) put it to work.

## 2 · The `bool` type

A boolean has exactly two possible values: `True` and `False`. Both are capitalised, and neither is in quotes.

````python
print(True)
print(False)
print(type(True))
````

Booleans are what Python gives back when you ask a yes/no question:

````python
print(10 > 3)
print(10 < 3)
print("a" in "banana")
````

You can store them in variables like any other value, and good names for them read like a question or a statement: `is_adult`, `has_license`, `game_over`.

````python
is_raining = True
has_umbrella = False
print(is_raining, has_umbrella)
````

## 3 · Comparison operators

| Operator | Question | Example | Result |
|---|---|---|---|
| `==` | equal? | `3 == 3` | `True` |
| `!=` | not equal? | `3 != 4` | `True` |
| `<` | less than? | `3 < 4` | `True` |
| `>` | greater than? | `3 > 4` | `False` |
| `<=` | less than or equal? | `3 <= 3` | `True` |
| `>=` | greater than or equal? | `3 >= 4` | `False` |

Remember from notebook 02: **`=` stores a value; `==` asks if two values are equal.**

````python
x = 7
print(x == 7)
print(x != 7)
print(x < 10)
print(x >= 8)
````

### Comparing different kinds of values

Numbers compare by size, and an `int` and a `float` can be compared freely:

````python
print(3 == 3.0)
print(2 < 2.5)
````

Strings compare for equality **exactly**, including case, and `"5"` is not `5`:

````python
print("hello" == "hello")
print("hello" == "Hello")
print("5" == 5)
````

Strings also compare by **dictionary order**, but "dictionary order" means by character code, in which *all capital letters come before all lowercase letters*:

````python
print("apple" < "banana")
print("Zebra" < "apple")      # capital Z (90) comes before lowercase a (97)
print(ord("Z"), ord("a"))
````

For case-insensitive comparisons, convert both sides first: `a.lower() == b.lower()`.

### Chained comparisons

Python lets you write maths-style ranges. `1 < x < 10` means "x is between 1 and 10":

````python
age = 15
print(13 <= age <= 19)       # is a teenager?
print(0 < age < 10)
````

## 4 · Combining conditions: `and`, `or`, `not`

Sometimes one comparison isn't enough. Three logical operators combine booleans:

- **`and`**: `True` only if **both** sides are true
- **`or`**: `True` if **at least one** side is true
- **`not`**: flips `True` to `False` and vice versa

Here are the **truth tables** (all possible inputs), printed from real code:

````python
print("A     B     A and B")
print(True, True, " ", True and True)
print(True, False, "", True and False)
print(False, True, "", False and True)
print(False, False, "", False and False)
````

````python
print("A     B     A or B")
print(True, True, " ", True or True)
print(True, False, "", True or False)
print(False, True, "", False or True)
print(False, False, "", False or False)
````

````python
print(not True)
print(not False)
````

In words: `and` is strict ("both"), `or` is generous ("either, or both"), `not` is a flip.

### In real code

````python
age = 25
has_ticket = True

can_enter = age >= 18 and has_ticket
print(can_enter)

gets_discount = age < 12 or age >= 65
print(gets_discount)

print(not has_ticket)
````

### Precedence

When you mix them, `not` is done first, then `and`, then `or`. Comparison operators bind tighter than all three, so `age >= 18 and has_ticket` works without brackets. Use parentheses when it isn't obvious:

````python
a, b, c = True, False, True

print(a or b and c)        # same as a or (b and c)
print((a or b) and c)
````

### Short-circuit evaluation

Python stops as soon as it knows the answer. `and` stops at the first `False`; `or` stops at the first `True`. The rest is never evaluated. This is useful for guarding against errors:

````python
print(False and 1 / 0)       # 1/0 would crash, but it never runs
print(True or 1 / 0)
````

## 5 · Truthiness

In conditions, Python doesn't insist on a `bool`; it accepts **any value** and decides whether it counts as true or false. `bool(x)` shows you the decision.

**The rule:** these are *falsy* (treated as `False`):

- `False`, `None`
- zero of any kind: `0`, `0.0`
- empty things: `""` (empty string), and later `[]`, `{}`, `()`, `set()`

**Everything else is *truthy*.**

````python
print(bool(0), bool(0.0), bool(""), bool(None))         # all False
print(bool(1), bool(-3), bool("0"), bool(" "), bool("False"))   # all True
````

Surprises to note: `"0"`, `" "` and `"False"` are truthy: they're non-empty strings, and Python doesn't read the text.

Truthiness lets you write short, natural tests like "is there any text?" or "is the count non-zero?":

````python
name = ""
print(bool(name))      # False: nothing was typed

count = 3
print(bool(count))     # True
````

### `and` / `or` return one of their operands

A subtle but useful detail: `and` and `or` don't always return `True`/`False`. They return **one of the two values**, whichever decided the result.

````python
print(0 or "default")        # 0 is falsy, so 'or' moves on to the second value
print("Ada" or "default")    # first value is truthy, so that's the answer
print(5 and 7)               # first is truthy, so the answer is the second
print(0 and 7)               # first is falsy, so the answer is that
````

The pattern `value or default` is a common way to supply a fallback, e.g. `name = input("Name: ") or "stranger"`. Use it sparingly, since it's easy to read wrongly.

## 6 · `None`: the absence of a value

`None` is Python's way of saying "nothing here". It has its own type, `NoneType`, and is falsy. You'll meet it as the placeholder for "no answer yet", and as the result of functions that don't produce a value (like `print()` itself!).

````python
result = None
print(result)
print(type(result))
print(bool(result))

answer = print("hi")     # print returns None
print(answer)
````

`None` is **not** the same as `0`, `""` or `False`: those are real values, which happen to be falsy. `None` means "no value at all".

## 7 · `is` versus `==`

- `==` asks: do these two have the **same value**?
- `is` asks: are these the **very same object** in memory?

For almost everything you'll write, you want `==`. The one place you should use `is` is when comparing with `None`:

````python
result = None
print(result is None)
print(result is not None)

other = 0
print(other is None)       # 0 is a value; it isn't None
print(other == 0)
````

Why `is None` rather than `== None`? There is only ever one `None`, so "is it that object?" is the exact question and can't be fooled by oddly-behaving objects.

Two equal values aren't necessarily the same object:

````python
a = int("1000")
b = int("1000")
print(a == b)      # same value
print(a is b)      # but two separate objects
````

> ⚠️ **Don't use `is` to compare numbers or strings.** Whether two equal numbers happen to be the same object depends on Python's internals (small numbers are reused; large ones usually aren't), and it can differ between versions. Use `==`. Level 9 covers the object model in depth.

## 8 · Common mistakes

### 1. `=` where `==` was meant

Assignment inside an expression isn't allowed. ⚠️ **Expected to fail.**

````python
x = 5
print(x = 5)
````

### 2. `x == 1 or 2`

This reads naturally in English ("x is 1 or 2") but Python reads it as `(x == 1) or 2`, and a non-zero number is always truthy:

````python
x = 5
print(x == 1 or 2)             # 2: truthy, so the whole thing is "true" no matter what x is
print(x == 1 or x == 2)        # what we meant: False
print(x in (1, 2))             # also what we meant (tuples are in Level 2)
````

Spell out the full comparison on each side of `or`/`and`.

### 3. Half a chained comparison

⚠️ **Expected to fail.**

````python
age = 15
print(age > 10 and < 20)
````

Each side of `and` must be a complete question: `age > 10 and age < 20` (or `10 < age < 20`).

### 4. Comparing numbers stored as text

Text compares character by character, so `"10"` comes *before* `"9"`:

````python
print("10" < "9")      # True: '1' comes before '9'
print(10 < 9)          # False
print(int("10") < int("9"))
````

This usually happens with values from `input()`. Convert before comparing (notebook 06).

### 5. Comparing incompatible types with `<`

You can ask `==` about anything, but the ordering operators need compatible types. ⚠️ **Expected to fail.**

````python
print("5" < 10)
````

### 6. Floating-point equality

As in notebook 03: `0.1 + 0.2 == 0.3` is `False`. Use `math.isclose()` or compare rounded values.

````python
import math

print(0.1 + 0.2 == 0.3)
print(math.isclose(0.1 + 0.2, 0.3))
````

### 7. Redundant comparisons with `True`

`if is_ready == True:` works but is clumsy: `is_ready` is already a boolean. Say `is_ready`, or `not is_ready`.

## 10 · Recap and what's next

- A **`bool`** is `True` or `False`. Comparisons (`== != < > <= >=`) produce them; `==` asks, `=` stores.
- Combine with **`and`**, **`or`**, **`not`** (precedence: `not`, then `and`, then `or`). Python **short-circuits**.
- Chained comparisons like `1 < x < 10` work as you'd hope. Strings compare by character code, so `"10" < "9"`.
- **Truthiness:** `False`, `None`, zero and empty values are falsy; everything else is truthy. `and`/`or` return one of their operands.
- **`None`** means "no value". Test for it with `is None`; use `==` for everything else.
