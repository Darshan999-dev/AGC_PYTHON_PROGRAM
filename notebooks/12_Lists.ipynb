## 1 · Why this matters

So far, to remember five test scores you would need five variables: `score1`, `score2`, ... and a program that handles ten scores would need ten. That doesn't scale, and you can't loop over variable names.

A **list** solves this: one variable holding an ordered collection of **items**. You can add to it, remove from it, sort it, loop over it, and ask how long it is. Shopping lists, high-score tables, the rows of a spreadsheet, the words of a sentence: they are all lists in a program.

Lists are also our first **mutable** type. *Mutable* means "can be changed in place after it is created". Strings (notebook 04) are **immutable**: you can't change a letter inside one. Lists can be changed, which is powerful, and which also brings one famous gotcha (section 8) that is worth learning early.

## 2 · Creating lists

## 3 · Indexing and slicing

Lists work exactly like strings here (notebook 04). Positions start at **0**, negative positions count from the end, and slices `[start:stop:step]` include `start` but stop *before* `stop`.

````python
fruit = ["apple", "banana", "cherry", "date", "elderberry"]

print(fruit[0])        # first
print(fruit[2])        # third
print(fruit[-1])       # last
print(fruit[-2])       # second to last
````

````python
print(fruit[1:3])      # positions 1 and 2
print(fruit[:2])       # the first two
print(fruit[2:])       # from position 2 to the end
print(fruit[::2])      # every second item
print(fruit[::-1])     # reversed
````

Note: **indexing** gives you one *item*; **slicing** gives you a new *list* (even if it has one item or none). Slices never raise an error for going past the end, but a single index does.

Here is the first difference from strings: because a list is mutable, you can **assign to a position** to replace an item.

````python
fruit[1] = "blueberry"
print(fruit)

fruit[0:2] = ["avocado", "banana"]      # you can even replace a whole slice
print(fruit)
````

## 4 · Asking questions about a list: `len`, `sum`, `min`, `max`, `in`

These built-in functions work on a whole list at once.

````python
scores = [72, 85, 90, 66, 78]

print("How many:", len(scores))
print("Total:   ", sum(scores))
print("Highest: ", max(scores))
print("Lowest:  ", min(scores))
print("Average: ", sum(scores) / len(scores))

print(90 in scores)           # membership test
print(100 in scores)
print(100 not in scores)
````

Two more lookups are handy: `.index(x)` gives the position of the first `x`, and `.count(x)` says how many times it appears. (Asking for `.index` of something that isn't there is an error, so check with `in` first if you are unsure.)

`sum` needs numbers; `min` and `max` need items that can be compared with each other (all numbers, or all strings).

````python
marks = [3, 1, 3, 2, 3]
print(marks.index(2))
print(marks.count(3))

words = ["pear", "apple", "fig"]
print(min(words), max(words))     # alphabetical order
````

## 5 · Looping over a list

This is the `for` loop from notebook 10, now with a list to walk through. The loop variable takes each **item** in turn.

````python
names = ["Ada", "Grace", "Alan"]

for name in names:
    print("Hello,", name)
````

When you also need the position, use `enumerate`; to walk two lists together, use `zip` (both from notebook 10). The accumulator patterns from notebook 11 work unchanged: `total = 0` then add inside the loop, or build up a **new list** starting from `[]`.

````python
for position, name in enumerate(names, start=1):
    print(position, name)

ages = [36, 85, 41]
for name, age in zip(names, ages):
    print(name, "is", age)

# Build a new list inside a loop: keep only the scores of 75 or more
scores = [72, 85, 90, 66, 78]
passed = []
for s in scores:
    if s >= 75:
        passed.append(s)
print(passed)
````

To *change* items while looping, loop over the **positions** and assign to `lst[i]`. Changing the loop variable alone does nothing to the list:

````python
prices = [10, 20, 30]

for p in prices:
    p = p * 2            # changes only the loop variable p, not the list
print(prices)

for i in range(len(prices)):
    prices[i] = prices[i] * 2
print(prices)
````

## 6 · Changing a list: the mutation methods

A **method** is a function that belongs to a value, called with a dot (like `"abc".upper()`). Lists have methods that **change the list itself**. Here is the toolbox, one at a time.

| Method | What it does |
|---|---|
| `lst.append(x)` | add `x` to the **end** |
| `lst.insert(i, x)` | put `x` at position `i`, shifting the rest right |
| `lst.extend(other)` | add **every item** of `other` to the end |
| `lst.remove(x)` | delete the **first** item equal to `x` |
| `lst.pop()` / `lst.pop(i)` | delete and **return** the last item / the item at `i` |
| `lst.sort()` | sort the list in place |
| `lst.reverse()` | reverse the list in place |

````python
todo = ["wash up", "email"]

todo.append("call mum")
print(todo)

todo.insert(0, "wake up")
print(todo)
````

`append` adds *one* item; `extend` adds *many*. The difference shows when you add a list:

````python
a = [1, 2]
a.append([3, 4])      # one item, which happens to be a list
print(a)

b = [1, 2]
b.extend([3, 4])      # the items 3 and 4
print(b)
````

`remove` deletes by **value**, `pop` deletes by **position** and hands the removed item back to you. (`del lst[i]` also deletes by position, and `lst.clear()` empties the list.)

````python
todo = ["wake up", "wash up", "email", "call mum"]

todo.remove("email")
print(todo)

last = todo.pop()           # remove and return the last item
print("popped:", last)
print(todo)

first = todo.pop(0)         # remove and return the item at position 0
print("popped:", first)
print(todo)
````

`sort()` and `reverse()` rearrange the list in place. `sort(reverse=True)` sorts from largest to smallest.

````python
nums = [5, 2, 9, 1]
nums.sort()
print(nums)

nums.sort(reverse=True)
print(nums)

nums.reverse()
print(nums)

words = ["banana", "Apple", "cherry"]
words.sort()
print(words)        # capital letters sort before lowercase ones
````

**Important habit:** these methods change the list and return `None`. So `nums = nums.sort()` would *destroy* your list (it would become `None`). Just call the method on its own line. We will see this under Common mistakes.

## 7 · `sorted()`: a sorted copy that leaves the original alone

`sorted(lst)` is a built-in *function* that returns a **new sorted list** and leaves the original untouched. Use `.sort()` when you want to change the list, and `sorted()` when you need the original too (or when you are sorting something that isn't a list, like a string or a `range`).

````python
nums = [5, 2, 9, 1]

ordered = sorted(nums)
print("ordered:", ordered)
print("nums:   ", nums)            # unchanged

print(sorted(nums, reverse=True))
print(sorted("python"))            # works on any sequence: a list of letters
print(nums[::-1])                  # a reversed copy, using a slice
````

**Preview (a later notebook explains it fully):** `sorted` and `.sort` accept a `key=` option naming a function that says *what to sort by*. For example `key=len` sorts words by length:

````python
words = ["banana", "fig", "cherry", "kiwi"]
print(sorted(words, key=len))
````

## 8 · Aliasing vs. copying: the first mutability gotcha

What do you think this prints?

````python
original = [1, 2, 3]
backup = original
original.append(4)

print("original:", original)
print("backup:  ", backup)
````

`backup` changed too! The line `backup = original` did **not** make a second list. It gave the *same* list a second name. Think of two name tags tied to one box:

```
original ──┐
           ├──►  [1, 2, 3, 4]
backup   ──┘
```

Two names for the same object is called an **alias**. We can confirm it with `is`, which asks "are these the very same object?" (a stricter question than `==`, which asks "do they hold equal items?").

````python
print(original is backup)       # same object?
print(original == backup)       # equal items?

other = [1, 2, 3, 4]
print(original == other)        # equal items...
print(original is other)        # ...but a different list
````

To get an independent **copy**, ask for one explicitly. Any of these create a new list with the same items:

````python
original = [1, 2, 3]

backup1 = original.copy()
backup2 = original[:]           # a slice of everything
backup3 = list(original)

original.append(4)
print("original:", original)
print("copies:  ", backup1, backup2, backup3)
print(original is backup1)
````

Why does Python work this way? Lists can be large, so assignment just attaches another name instead of silently duplicating all the items. It's efficient, but it means **mutating through one name is visible through every alias**.

This also explains a common surprise when you pass a list somewhere to be changed: everyone sharing that list sees the change. And it doesn't happen with numbers or strings, because those are immutable: there's no way to change them in place, so sharing is harmless.

````python
x = 5
y = x
x = x + 1          # makes a NEW int and rebinds x; y is unaffected
print(x, y)

s = "cat"
t = s
s = s + "s"        # a NEW string; t is unaffected
print(s, t)
````

## 9 · Nested lists (2D data)

A list's items can themselves be lists. A list of rows is the natural way to store a **grid** or a table: spreadsheet rows, a game board, an image of pixels, a matrix.

````python
grid = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]

print(grid)
print(len(grid))          # number of rows
print(len(grid[0]))       # number of items in the first row
````

Use **two indexes**: `grid[row][col]`. The first picks a row (itself a list), the second picks an item in that row.

````python
print(grid[0])          # a whole row
print(grid[1][2])       # row 1, column 2
print(grid[-1][-1])     # bottom-right corner

grid[2][0] = 70         # change one cell
print(grid)
````

Nested loops (notebook 11) visit every cell: the outer loop goes over rows, the inner loop over the items in that row.

````python
for row in grid:
    for value in row:
        print(f"{value:3}", end="")
    print()

total = 0
for row in grid:
    total += sum(row)           # sum() of each row
print("Total:", total)
````

### Careful: building a grid

`[0] * 3` is a fine way to make a row. But `[[0] * 3] * 3` does **not** make three separate rows: it repeats the *same* row object three times (an alias again!). The right way is to build each row separately, in a loop:

````python
bad = [[0] * 3] * 3
bad[0][0] = 1
print(bad)           # all three rows changed!

good = []
for r in range(3):
    good.append([0] * 3)      # a fresh row each time round
good[0][0] = 1
print(good)
````

Copies of nested lists are also tricky: `grid.copy()` copies the *outer* list, but the inner row lists are still shared (a **shallow** copy).

````python
grid = [[1, 2], [3, 4]]
shallow = grid.copy()
shallow[0][0] = 99
print(grid)             # grid changed too!

safe = []
for row in grid:
    safe.append(row.copy())     # copy every row
safe[1][1] = -1
print(grid, safe)
````

(There is a ready-made tool for this, `copy.deepcopy`, that you will meet when we learn about modules. For now, copying row by row is fine.)

## 10 · Common mistakes

These are all errors real learners hit. The cells that crash are meant to.

**1. Off-by-one: the last index is `len - 1`.** ⚠️ **Expected to fail.**

````python
colors = ["red", "green", "blue"]
print(colors[3])
````

Valid positions are `0, 1, 2` here. Use `colors[-1]` for "the last one" and `range(len(colors))` for all valid positions.

**2. `remove` of something that isn't there, and `pop` from an empty list.** ⚠️ **Expected to fail.**

````python
colors = ["red", "green", "blue"]
colors.remove("yellow")
````

⚠️ **Expected to fail.**

````python
empty = []
empty.pop()
````

Guard with `if "yellow" in colors:` or `if empty:` first. (In Level 3 you'll learn another way, `try`/`except`.)

**3. Saving the result of a method that returns `None`.**

````python
nums = [3, 1, 2]
nums = nums.sort()      # sort() changes the list and returns None
print(nums)
````

````python
nums = [3, 1, 2]
result = nums.append(4)
print(result)           # also None. To keep the list, just call nums.append(4)
````

Remember: methods that change a list in place return `None`. `sorted(nums)` is the one that returns a new list.

**4. Removing items while looping over the same list.** The loop walks by position, so when you delete an item, the next one slides into the spot just visited and gets skipped:

````python
nums = [1, 2, 2, 3, 4, 4, 5]
for n in nums:
    if n % 2 == 0:
        nums.remove(n)
print(nums)             # not all the even numbers are gone!

# Safer: build a new list of the items you want to keep
nums = [1, 2, 2, 3, 4, 4, 5]
kept = []
for n in nums:
    if n % 2 != 0:
        kept.append(n)
print(kept)
````

**5. Thinking `=` copies** (section 8), and **repeating a list of lists with `*`** (section 9). Both create aliases when you wanted separate lists.

**6. `append` vs `extend` vs `+`.** `a.append([3, 4])` adds one nested list. `a.extend([3, 4])` adds two items. `a + [3, 4]` builds a *new* list and leaves `a` alone.

## 12 · Recap and what's next

- A **list** is an ordered, **mutable** collection of items: `[1, "two", 3.0]`. Index and slice it like a string, but you may also assign to positions.
- `len`, `sum`, `min`, `max` and `in` ask questions about a list. `for item in lst` walks through it; `enumerate` and `zip` help with positions and pairs.
- Methods that change a list in place and return `None`: `append`, `insert`, `extend`, `remove`, `pop` (this one returns the removed item), `sort`, `reverse`. `sorted()` returns a new list instead.
- `b = a` makes an **alias**, not a copy. Use `a.copy()`, `a[:]` or `list(a)` for a copy, and remember that copies of nested lists are shallow.
- A list of lists models a grid: `grid[row][col]`. Build rows separately, never with `[[0] * 3] * 3`.
