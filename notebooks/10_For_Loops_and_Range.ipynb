## 1 · Why this matters

A `while` loop makes you manage the counter yourself: initialise, test, update. That's three places to make a mistake. But most loops fall into one of two simple families: "do this **N times**" and "do this **for each item** in something". Python has a dedicated loop for both, the `for` loop, and it does the bookkeeping for you. You'll write far more `for` loops than `while` loops.

## 2 · Looping over a string

```
for variable in sequence:
    body
```

The loop takes the items of the sequence one at a time, assigns each to the **loop variable**, and runs the body once per item. A string is a sequence of characters:

````python
for letter in "Python":
    print(letter)
````

You choose the name of the loop variable (`letter` here). It's an ordinary variable that happens to be reassigned on every repetition. Pick a name that says what each item *is*.

````python
total_letters = 0
for character in "hello world":
    total_letters += 1
print(total_letters)
````

(There's no need to count like this, since `len()` does it. We're showing the loop mechanics.)

### Looping over words

The `.split()` method (notebook 05) gives a list of words, and a `for` loop walks over its items just like the characters of a string:

````python
sentence = "the quick brown fox"
for word in sentence.split():
    print(word.upper())
````

## 3 · `range()`: a sequence of numbers

To repeat something N times, or to loop over numbers, use `range()`.

| Call | Produces |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 |
| `range(5, 0, -1)` | 5, 4, 3, 2, 1 |

Just like slicing in notebook 04: the **start** is included, the **stop** is **excluded**, and the third number is the **step**. With one number, it's the stop (start defaults to 0).

````python
for i in range(5):
    print(i, end=" ")
print()

for i in range(2, 6):
    print(i, end=" ")
print()

for i in range(0, 10, 2):
    print(i, end=" ")
print()

for i in range(5, 0, -1):
    print(i, end=" ")
print()
````

To *see* a range without a loop, wrap it in `list()` (more on lists in Level 2):

````python
print(list(range(5)))
print(list(range(1, 11)))
print(list(range(10, 0, -3)))
print(len(range(100)))
````

**The easy way to remember:** `range(n)` gives exactly `n` numbers starting at 0, so `for _ in range(3)` repeats something three times. `range(a, b)` gives `b - a` numbers.

If you don't need the loop variable at all, the convention is to call it `_`:

````python
for _ in range(3):
    print("Hip hip hooray!")
````

## 4 · `for` or `while`?

| Use `for` when... | Use `while` when... |
|---|---|
| you know how many times | you don't know how many times |
| you're going through every item of something | you're waiting for something to become true |
| *"for each..."*, *"N times"* | *"until..."*, *"as long as..."* |

Examples: printing a times table (`for`); asking until a valid answer (`while`); adding up the numbers 1 to 100 (`for`); halving a number until it's below 1 (`while`).

A `for` loop can't run forever on a `range` or a string, since they're finite, which is one reason they're safer.

## 5 · Loop patterns

Most loops are one of a handful of patterns. Learn to recognise them.

### Accumulate: sum and product

````python
total = 0
for n in range(1, 101):
    total += n
print("Sum 1..100:", total)

product = 1
for n in range(1, 6):
    product *= n
print("5! =", product)
````

Notice again: **initialise before the loop** (`0` for sums, `1` for products), update inside, use the result after.

### Count: how many items match?

````python
count = 0
for letter in "mississippi":
    if letter == "s":
        count += 1
print("s appears", count, "times")
````

### Build a string

````python
shouted = ""
for letter in "hello":
    shouted += letter.upper() + "!"
print(shouted)

backwards = ""
for letter in "stressed":
    backwards = letter + backwards      # put each new letter at the front
print(backwards)
````

### Search: is there one?

Use a **flag**: a boolean that starts `False` and is switched to `True` if the loop finds the thing.

````python
found = False
for letter in "hello":
    if letter == "z":
        found = True
        break
print("Contains z?", found)
````

`break` and `continue` work in `for` loops just as in `while` loops. A `for` loop may also have an `else` block that runs if it finishes without a `break`:

````python
for letter in "hello":
    if letter == "z":
        print("found a z")
        break
else:
    print("no z in hello")
````

## 6 · `enumerate()`: position and item

Sometimes you want each item *and* its position. Beginners often reach for `range(len(...))`, but there's a better way, `enumerate()`, which hands you `(position, item)` pairs:

````python
for i, letter in enumerate("cab"):
    print(i, letter)
````

`for i, letter in ...` unpacks each pair into two variables, the same trick as `a, b = 1, 2` in notebook 02. By default the counting starts at 0; give `start=1` to count like a human:

````python
for place, name in enumerate("XYZ", start=1):
    print(f"{place}. {name}")
````

Compare with the clumsy way, which also works:

````python
word = "cab"
for i in range(len(word)):
    print(i, word[i])
````

Same result; `enumerate` is shorter and avoids indexing mistakes.

## 7 · `zip()`: walk two sequences together

````python
letters = "abc"
digits = "123"

for letter, digit in zip(letters, digits):
    print(letter + digit)
````

`zip` pairs up the first items, the second items, and so on. It stops when the **shorter** one runs out:

````python
for x, y in zip("abcde", "12"):
    print(x, y)
````

## 8 · Common mistakes

### 1. `range` stops before the stop value

`range(1, 10)` ends at 9. To include 10 write `range(1, 11)`.

````python
for i in range(1, 10):
    print(i, end=" ")
print()
````

### 2. Wrong direction

A positive step can't go down, and a negative step can't go up; the loop just doesn't run:

````python
for i in range(10, 0):
    print("never printed")
print(list(range(10, 0)))       # empty
print(list(range(10, 0, -1)))   # counting down needs a negative step
````

### 3. `range` needs whole numbers

⚠️ **Expected to fail.**

````python
for i in range(2.5):
    print(i)
````

### 4. Looping over a number

A number isn't a sequence. Maybe you wanted `range(5)`, or the digits of `str(12345)`. ⚠️ **Expected to fail.**

````python
for digit in 12345:
    print(digit)
````

````python
for digit in str(12345):
    print(digit)
````

### 5. Putting the accumulator's start *inside* the loop

If you reset the total on every repetition, you only ever keep the last item:

````python
for n in range(1, 4):
    total = 0           # oops: reset every time
    total += n
print(total)            # 3, not 6
````

### 6. Indentation decides what's in the loop

Compare these. In the first, `print` is inside the loop (printed 3 times); in the second it's after the loop (printed once):

````python
total = 0
for n in range(1, 4):
    total += n
    print("running total:", total)

print("---")

total = 0
for n in range(1, 4):
    total += n
print("final total:", total)
````

### 7. Changing the loop variable has no effect on the loop

Each repetition the loop assigns the next item to the variable, wiping out whatever you did to it:

````python
for i in range(3):
    print("before:", i, end=" ")
    i = i * 10
    print("after:", i)
````

## 10 · Recap and what's next

- **`for item in sequence:`** runs the body once per item; Python does the bookkeeping for you. You can loop over strings and over `.split()` results.
- **`range(stop)`**, **`range(start, stop)`**, **`range(start, stop, step)`**: the start is included and the stop is excluded. A negative step counts down.
- Use `for` when you know the count or have a collection to go through; use `while` when you're waiting for a condition.
- Patterns: **accumulate** (sum, product, build a string), **count**, **search with a flag**, **filter with `if`**.
- **`enumerate(x)`** gives `(position, item)` pairs; **`zip(a, b)`** pairs up items from two sequences.
