## 2 · Nested loops

A loop inside another loop is a **nested loop**. For **every single repetition of the outer loop**, the inner loop runs through *all* of its repetitions.

````python
for i in range(3):
    for j in range(2):
        print("i =", i, " j =", j)
````

Read the output carefully: `i` stays the same while `j` goes through `0, 1`, then `i` advances and `j` starts over. It's like a clock: the seconds hand goes round for every tick of the minutes hand.

**How many times does the inner body run?** `3 × 2 = 6`: the outer count times the inner count.

### Printing a grid

To print rows of characters, the inner loop prints one row's cells on a single line (`end=""`), and then an empty `print()` ends the line before the outer loop moves on.

````python
for row in range(3):
    for col in range(5):
        print("#", end="")
    print()                # finish the row
````

The inner loop draws a row; the outer loop repeats the row. This "outer = rows, inner = columns" structure is how almost all grids are drawn.

### The inner range can depend on the outer variable

That makes a triangle: row 1 has one `#`, row 2 has two, and so on.

````python
for row in range(1, 6):
    for col in range(row):
        print("#", end="")
    print()
````

(You could do this with `print("#" * row)`, but spelling it out with loops teaches the mechanism you'll need for less regular shapes.)

### A table

The multiplication table: rows and columns are both numbers, and the formatting `{value:4}` pads each to width 4 so the columns line up:

````python
for row in range(1, 6):
    for col in range(1, 6):
        print(f"{row * col:4}", end="")
    print()
````

### `break` only leaves the *inner* loop

`break` exits the loop it's directly inside. To stop *both*, use a flag, or put the code somewhere you can `return` from (functions come in Level 2).

````python
found = False
for a in range(1, 10):
    for b in range(1, 10):
        if a * b == 42:
            found = True
            break           # leaves the inner loop only
    if found:
        break               # leaves the outer loop

print("First pair with product 42:", a, b)
````

## 3 · Loop patterns

You've already used these in notebooks 9 and 10; here they're named so you can recognise them.

### Accumulate

Combine every item into one result: a sum, a product, a joined string.

### Filter

Act on only the items that satisfy a condition (`if` inside `for`):

````python
total_even = 0
for n in range(1, 21):
    if n % 2 == 0:
        total_even += n
print(total_even)
````

### Find the first

Look until you find a match, then stop with `break`:

````python
for n in range(1, 100):
    if n * n > 500:
        print("First n with n squared over 500:", n)
        break
````

### Track the minimum or maximum

Keep the best value seen so far, and replace it when something better comes along. The question is how to start: pick something that any real item will beat, or use `None` to mean "nothing seen yet".

````python
sentence = "the quick brown fox jumped over the lazy dogs"

longest = ""
for word in sentence.split():
    if len(word) > len(longest):     # '>' keeps the FIRST of equal-length words
        longest = word
print("Longest word:", longest)

shortest = None
for word in sentence.split():
    if shortest is None or len(word) < len(shortest):
        shortest = word
print("Shortest word:", shortest)
````

Two things to note. With `>` (not `>=`) the *first* of several equally good items wins. And starting `shortest` at `0` or `""` would be a bug: nothing is shorter than that, so it would never be replaced.

### Count with a condition and compute a rate

````python
vowels = 0
letters = 0
for ch in "Programming is fun":
    if ch.isalpha():
        letters += 1
        if ch.lower() in "aeiou":
            vowels += 1
print(f"{vowels} vowels out of {letters} letters = {vowels / letters:.0%}")
````

### All pairs

Nested loops are the natural tool for "every combination of two things". To avoid listing both `(2, 9)` and `(9, 2)`, start the inner loop after the outer variable:

````python
target = 10
for a in range(1, 10):
    for b in range(a + 1, 10):
        if a + b == target:
            print(a, "+", b, "=", target)
````

## 4 · Choosing between `for` and `while`

A quick decision guide:

1. **Do I have a collection (or a count) to go through?** Use `for`.
2. **Am I waiting for a condition that depends on what happens inside the loop** (user input, a calculation that converges, a game state)? Use `while`.
3. **Is it "keep going until..."?** `while`. **"For each..."?** `for`.

You can always rewrite one as the other, but one will read more naturally. If you notice you're manually initialising and incrementing a counter in a `while`, a `for` with `range` is usually clearer.

Also be aware of cost: nested loops **multiply**. A 1,000 × 1,000 grid is a million repetitions (fine); 100,000 × 100,000 is ten billion (not fine). This is the start of thinking about efficiency, which Level 9 covers.

## 5 · Text-based game boards

Nested loops are also how simple text games draw their screen. Here's a tic-tac-toe style board, using the rule "X when `row + col` is even, otherwise O":

````python
for row in range(3):
    line = ""
    for col in range(3):
        mark = "X" if (row + col) % 2 == 0 else "O"
        line += mark
        if col < 2:
            line += " | "
    print(line)
    if row < 2:
        print("--+---+--")
````

Notice the conditional expression and the `if col < 2` trick for separators between cells but not after the last one.

## 6 · Common mistakes

### 1. Using the same variable name in both loops

The inner loop's variable overwrites the outer one. Use different names (`row`/`col`, `i`/`j`).

````python
for i in range(3):
    for i in range(2):      # reuses i!
        pass
    print("outer i is now", i)    # always 1 (the inner loop's last value)
````

### 2. Forgetting the `print()` that ends each row

Without it the whole grid appears on one line:

````python
for row in range(3):
    for col in range(4):
        print("#", end="")
print()
````

### 3. Wrong indentation

Which loop does a line belong to? Compare:

````python
for i in range(3):
    for j in range(3):
        print(j, end=" ")
    print("|", i)        # runs once per i (after the inner loop)

print()

for i in range(3):
    for j in range(3):
        print(j, end=" ")
        print("|", i)    # runs once per j!
````

### 4. Starting a min/max at the wrong value

Starting `smallest = 0` for numbers that are all positive means `smallest` never changes. Start with `None` or with the first item.

### 5. Off-by-one in `range`

Remember that `range(5)` is 0 to 4 and `range(1, 5)` is 1 to 4. Print a tiny case, say `range(3)`, and check by eye.

### 6. Expecting `break` to leave every loop

It leaves only the innermost loop. Use a flag, as above.

### 7. Doing too much work in the inner loop

If something doesn't depend on the inner variable, compute it in the outer loop instead of repeating it. We'll see this more in Level 9.
