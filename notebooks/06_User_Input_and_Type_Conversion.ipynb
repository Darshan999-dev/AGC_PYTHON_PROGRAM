## 3 · `input()` always returns a string

Here is the catch: whatever the user types, you get a `str`. Even if they type digits. Let's see what happens when we try arithmetic on the result.

````python
age = input("How old are you? ")
print(age)
print(type(age))
````

The value looks like a number, but it's the **text** `"30"`. A string of digits can't be added to a number. ⚠️ **Expected to fail.**

````python
age = input("How old are you? ")
print("Next year you will be", age + 1)
````

Worse, if you do something that *looks* like it should work:

````python
a = input("First number: ")
b = input("Second number: ")
print(a + b)
````

`"3" + "4"` is `"34"`: `+` joined two strings. No error, but the wrong answer, which is more dangerous than an error. The fix is to **convert**.

## 4 · Type conversion

Python can convert a value from one type to another using the **name of the type** as a function:

| Function | Converts to... | Examples |
|---|---|---|
| `int(x)` | whole number | `int("42")` → `42`, `int(3.9)` → `3` |
| `float(x)` | decimal number | `float("3.14")` → `3.14`, `float(5)` → `5.0` |
| `str(x)` | text | `str(42)` → `"42"`, `str(3.0)` → `"3.0"` |
| `bool(x)` | `True` or `False` | `bool(0)` → `False`, `bool("hi")` → `True` |

Conversion creates a **new** value; the original is unchanged.

````python
print(int("42") + 1)
print(float("3.14") * 2)
print(str(42) + "!")
print(int(3.9), int(-3.9))      # int() chops the decimal part (towards zero)
print(float(5))
print(str(3.0))
````

### Converting what the user typed

Now we can fix the age program. Convert as soon as you have the text:

````python
age = int(input("How old are you? "))
print("Next year you will be", age + 1)
````

`int(input(...))` is a very common one-liner: the inner call `input(...)` runs first and returns a string, then `int(...)` converts it. If you find it hard to read, split it into two lines; they do exactly the same:

````python
age_text = input("How old are you? ")
age = int(age_text)
print("Next year you will be", age + 1)
````

Use `float()` instead of `int()` when the number might have a decimal part:

````python
height = float(input("Your height in metres: "))
print("In centimetres:", height * 100)
````

### Spaces are fine, but junk is not

`int()` and `float()` ignore spaces around the number, which helps when users type sloppily:

````python
print(int("  42  "))
print(float(" 3.5\n"))
````

### Truthiness: `bool()`

`bool()` turns any value into `True` or `False`. Roughly: "empty" and "zero" values are `False`, everything else is `True`. (We'll formalise this in notebook 07.)

````python
print(bool(0), bool(1), bool(-5))
print(bool(""), bool("hello"))
print(bool("False"))        # a non-empty string, so True! Python doesn't read the text
````

## 5 · When a conversion fails

Not every string is a number. If the text can't be converted, Python raises a **`ValueError`**. ⚠️ **Expected to fail.**

````python
print(int("hello"))
````

Some failures that surprise people:

| Code | Result | Why |
|---|---|---|
| `int("3.5")` | `ValueError` | `int()` won't accept a decimal point in text; use `int(float("3.5"))` |
| `int("")` | `ValueError` | an empty string is not a number |
| `int("1,000")` | `ValueError` | the comma isn't allowed |
| `int("12 apples")` | `ValueError` | nothing but the number is allowed |
| `float("1e3")` | `1000.0` | scientific notation is fine |
| `int("-5")` | `-5` | a leading minus is fine |

⚠️ **Expected to fail.** Try the surprising first one:

````python
print(int("3.5"))
````

````python
print(int(float("3.5")))     # convert in two steps
````

### What do we do about bad input?

## 6 · Structuring an interactive program

Nearly every small program follows the same four steps:

1. **Input**: ask for things (as text)
2. **Convert**: turn the text into the types you need
3. **Compute**: do the calculation
4. **Output**: show the result nicely

Keeping these steps separate makes a program easy to read and easy to fix. Here's a temperature converter:

````python
# 1. input
text = input("Temperature in Celsius: ")

# 2. convert
celsius = float(text)

# 3. compute
fahrenheit = celsius * 9 / 5 + 32

# 4. output
print(f"{celsius}°C is {fahrenheit:.1f}°F")
````

## 7 · Common mistakes

### 1. Forgetting to convert

`input("Number: ") + 1` fails; `input("A: ") + input("B: ")` quietly concatenates. Whenever you ask for a number, convert it.

### 2. Putting `int` in the wrong place

`int(input)` (no brackets after `input`) tries to convert the function itself. ⚠️ **Expected to fail.**

````python
age = int(input)
````

You need `int(input("..."))`: input is **called**, and its result is converted.

### 3. Converting at the wrong end

The conversion happens on the *string you received*. Writing `input(int("Age: "))` converts the prompt. ⚠️ **Expected to fail.**

````python
age = input(int("Age: "))
````

### 4. Using `int()` for something that might have decimals

If the user types `1.75` and you call `int()` on it, you get a `ValueError`. Use `float()`, or decide to accept whole numbers only.

### 5. Expecting `bool("False")` to be `False`

`bool()` of any non-empty string is `True`, whatever it says. To interpret the *text*, compare it: `answer.lower() == "yes"`.

````python
answer = "No"
print(bool(answer))                    # True: it's a non-empty string
print(answer.lower() == "yes")         # False: this is what we actually meant
````

### 6. Forgetting the output formatting

`print("Total:", 3 * 1.1)` shows `3.3000000000000003` (float noise). Format money and measurements with an f-string like `{total:.2f}`.
