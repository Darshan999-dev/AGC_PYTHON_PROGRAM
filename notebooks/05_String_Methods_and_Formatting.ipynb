## 1 · Why this matters

Real text is messy: extra spaces, ALL CAPS, inconsistent punctuation. Before a program can do anything useful with it, it usually has to **clean it up**. And once it has results, it has to **present** them in a readable way: aligned columns, prices with two decimals, thousands separators.

Python strings come with dozens of built-in tools for exactly this. In this notebook you meet the ones you'll use every week.

## 2 · Methods

A **method** is a function that belongs to a value. You call it by writing the value, a dot, the method name and brackets:

```
value.method(arguments)
```

````python
message = "hello, world"
print(message.upper())
print(message.title())
print(message.count("o"))
````

Think of the dot as "ask this string to...": `message.upper()` means "ask `message` to give you an uppercase version of itself".

### Methods return new strings

Remember from notebook 04 that strings are **immutable**. A method never changes the original; it hands you a new string. So if you want to keep the result you have to store it:

````python
shout = "quiet"
shout.upper()
print(shout)             # unchanged!

shout = shout.upper()    # store the new string
print(shout)
````

## 3 · The methods you'll use most

### Changing case

| Method | Effect | `"hELLo wORLD"` becomes |
|---|---|---|
| `.upper()` | all capitals | `"HELLO WORLD"` |
| `.lower()` | all lowercase | `"hello world"` |
| `.title()` | Each Word Capitalised | `"Hello World"` |
| `.capitalize()` | first letter only | `"Hello world"` |
| `.swapcase()` | flip every letter | `"HellO WorlD"` |

````python
text = "hELLo wORLD"
print(text.upper())
print(text.lower())
print(text.title())
print(text.capitalize())
````

### Removing unwanted whitespace

`.strip()` removes spaces, tabs and newlines from **both ends** (not the middle). `.lstrip()` and `.rstrip()` do just the left or right end. You can also pass the characters to strip.

````python
messy = "   too much space   "
print("[" + messy.strip() + "]")
print("[" + messy.lstrip() + "]")
print("[" + messy.rstrip() + "]")
print("xxhixx".strip("x"))
````

### Replacing

`.replace(old, new)` swaps **every** occurrence:

````python
print("a-b-c".replace("-", "+"))
print("the cat and the hat".replace("the", "a"))
print("Hello".replace("l", ""))      # replacing with "" deletes
````

### Searching and testing

| Method | What it tells you |
|---|---|
| `.find(sub)` | index of the first occurrence, or `-1` if absent |
| `.index(sub)` | same, but raises an error if absent |
| `.count(sub)` | how many (non-overlapping) times it appears |
| `.startswith(sub)` / `.endswith(sub)` | `True` or `False` |
| `.isdigit()` / `.isalpha()` / `.isalnum()` | is every character a digit / a letter / a letter-or-digit? |

````python
sentence = "Mississippi"
print(sentence.find("ss"))
print(sentence.find("z"))
print(sentence.count("s"))
print(sentence.count("ss"))
print(sentence.startswith("Miss"))
print(sentence.endswith("pi"))
print("12345".isdigit(), "12a45".isdigit())
````

`.find()` returning `-1` for "not found" is a classic convention. Be careful because `-1` is also a valid index!

### Splitting and joining

`.split()` chops a string into pieces and gives you back a **list** (an ordered collection of items; Level 2 is all about them). With no argument it splits on whitespace; with an argument it splits at that separator.

````python
words = "the quick brown fox".split()
print(words)
print(len(words))          # len() works on lists too: the number of items

print("2026-10-18".split("-"))
print("a,b,,c".split(","))
````

You can pick items out of a list with `[ ]`, exactly as with characters in a string:

````python
parts = "2026-10-18".split("-")
print(parts[0])
print(parts[-1])
````

`.join()` is the opposite of `.split()`: it glues a list of strings into one string, with the string you call it on placed **between** them. This looks backwards at first: the *separator* is the string you call `.join` on.

````python
words = ["the", "quick", "brown", "fox"]
print(" ".join(words))
print("-".join(words))
print("".join(words))
````

Because each method returns a string, you can call another method straight on the result:

## 4 · f-strings: strings with values inside

Glueing strings with `+` and converting numbers with `str()` gets ugly fast:

```python
"Hello, " + name + "! You are " + str(age) + " years old."
```

An **f-string** puts the values right where they belong. Put an `f` before the opening quote and write variables or expressions inside `{ }`:

````python
name = "Ada"
age = 36

print(f"Hello, {name}! You are {age} years old.")
````

Anything inside the braces is evaluated as Python. You can do arithmetic, call methods and call functions:

````python
print(f"Next year you will be {age + 1}.")
print(f"{name.upper()} has {len(name)} letters.")
print(f"{age} squared is {age ** 2}.")
````

### Debugging shortcut: `=`

A trailing `=` inside the braces prints both the expression *and* its value. Wonderful for quick debugging:

````python
x = 3
y = 4
print(f"{x=}, {y=}, {x + y=}")
````

### Literal braces

To print an actual brace, double it:

````python
print(f"A set looks like {{1, 2, 3}}, and x is {x}")
````

## 5 · Formatting values

After the expression you can add a colon `:` and a **format specification** that controls how it's displayed. This is the most powerful part of f-strings.

### Decimal places and thousands separators

````python
pi = 3.14159265
print(f"{pi:.2f}")          # 2 decimal places
print(f"{pi:.0f}")
print(f"{1234567:,}")       # thousands separator
print(f"{1234567.891:,.2f}")
print(f"{0.4567:.1%}")      # percentage
````

The pattern `.2f` means "a float with 2 digits after the point". Use `,` for thousands separators and `%` to multiply by 100 and add a percent sign.

### Padding and alignment

`{value:>10}` means "right-align in a field 10 characters wide". The alignment characters are:

| Symbol | Alignment |
|---|---|
| `<` | left |
| `>` | right |
| `^` | centre |

````python
print(f"[{'left':<10}]")
print(f"[{'right':>10}]")
print(f"[{'mid':^10}]")
print(f"[{'fill':*^10}]")      # a character before the symbol is used as filler
print(f"[{42:>8}]")
print(f"[{3.14159:>8.2f}]")
````

Numbers are right-aligned by default and strings are left-aligned, so you only need the symbols when you want something different.

### Zero padding, signs and number bases

````python
print(f"{7:03d}")      # zero-pad to width 3
print(f"{42:06d}")
print(f"{5:+d}")       # always show the sign
print(f"{10:b}")       # binary
print(f"{255:x}")      # hexadecimal
````

### A table

Alignment lets you build tidy tables. The width can itself come from a variable using nested braces:

````python
width = 12
print(f"{'Item':<{width}}{'Price':>8}")
print("-" * (width + 8))
print(f"{'Coffee':<{width}}{3.5:>8.2f}")
print(f"{'Sandwich':<{width}}{6.25:>8.2f}")
````

### Other ways you will see in older code

Before f-strings (Python 3.6) there were two other styles. You don't need them, but you'll meet them in other people's code:

````python
print("{} is {} years old".format("Ada", 36))     # str.format()
print("%s is %d years old" % ("Ada", 36))         # the old % style
````

## 6 · Unicode and emoji

Python strings are **Unicode**: every character in every writing system (plus symbols and emoji) has a unique number, called a **code point**. That's why Python happily handles `"héllo"`, `"こんにちは"` and `"🐍"`.

````python
print("héllo", "こんにちは", "🐍")
print(ord("A"), ord("é"), ord("🐍"))     # character -> code point number
print(chr(97), chr(233), chr(128013))    # code point number -> character
print("\u00e9", "\N{snowman}")      # write characters by number (\u) or by name (\N)
````

### Careful: one *thing you see* is not always one character

````python
print(len("é"))          # one code point
print(len("é"))    # 'e' followed by a combining accent: looks identical, length 2!
print(len("👍"))
print(len("👍🏽"))        # thumbs-up + skin-tone modifier = two code points
````

`len()` counts **code points**, not what a human perceives as characters. For everyday English text they're the same; for accents and emoji sometimes they aren't. Case conversion can also surprise you:

````python
print("ß".upper())       # the German sharp s becomes two letters
````

Computers store text as bytes using an **encoding**; the standard is `UTF-8`. You'll see that when we read and write files in Level 3.

````python
print("é".encode("utf-8"))
````

## 7 · Common mistakes

### 1. Forgetting to store the result

Methods never change the original string.

````python
name = "ada"
name.title()
print(name)         # still "ada"
````

### 2. Forgetting the brackets

`name.title` without `()` isn't a call; it just refers to the method.

````python
print(name.title)
print(name.title())
````

### 3. Expecting `.strip()` to remove inner spaces

````python
print("  a  b  ".strip())            # the spaces between a and b stay
print("  a  b  ".replace(" ", ""))   # to remove all of them, replace
````

### 4. `.replace()` replaces everything, and everything is case-sensitive

````python
print("Banana".replace("a", "o"))        # every lowercase a
print("Banana".count("b"))               # 0! the b is capital
print("Banana".lower().count("b"))       # 1
````

### 5. Forgetting the `f`

Without the `f` prefix the braces are just characters:

````python
name = "Ada"
print("Hello, {name}!")
print(f"Hello, {name}!")
````

### 6. `join` needs strings, and is called on the separator

⚠️ **Expected to fail.**

````python
print(", ".join([1, 2, 3]))
````

`join` only joins strings. Convert first, e.g. `", ".join(["1", "2", "3"])`. (In Level 4 you'll learn a one-liner that converts a whole list.)

### 7. `.split(" ")` versus `.split()`

With multiple spaces in a row, `.split(" ")` gives empty strings but `.split()` doesn't:

````python
print("a  b".split(" "))
print("a  b".split())
````

## 9 · Recap and what's next

- A **method** is called with `value.method()`. Strings are immutable, so methods **return new strings**; store the result if you want to keep it.
- Cleaning: `strip`, `lower`, `upper`, `title`, `replace`. Searching: `find`, `count`, `startswith`, `endswith`. Splitting and joining: `split` and `" ".join(...)`.
- **f-strings** (`f"... {value} ..."`) embed expressions; the `:spec` after a colon controls decimals (`.2f`), separators (`,`), alignment (`<`, `>`, `^`), padding and more.
- Python strings are **Unicode**. `len()` counts code points, which isn't always what a human sees as one character.
