## 1 · Why this matters

A list finds things by **position**: "the third item". But much of the information we handle is found by **name**: the phone number *of Ada*, the price *of milk*, the number of times the word *"the"* appears. With a list you would have to search through everything, or remember which position belongs to which name.

A **dictionary** (type `dict`) solves this. It stores **key-value pairs**: you give it a key, and it hands back the value. Like a real dictionary, you look up a word (the key) to get its meaning (the value), without reading from page one.

Dictionaries are everywhere in Python: configuration settings, records read from files and web services, counters, caches, and, later in the course, the objects of classes themselves are built on them.

## 3 · Looking things up

Put the key in square brackets, just like an index. The difference: for a dictionary, the thing in brackets is a **key**, not a position.

```python
ages = {"Ada": 36, "Grace": 45, "Linus": 28}

print(ages["Grace"])
print("Ada is", ages["Ada"], "years old")
```

Asking for a key that is not there is an error (a `KeyError`): ⚠️ **Expected to fail.**

```python
print(ages["Alan"])
```

To check first, use `in`. For a dictionary it tests the **keys** (not the values):

```python
print("Ada" in ages)
print("Alan" in ages)
print("Alan" not in ages)
print(36 in ages)       # 36 is a value, not a key

if "Linus" in ages:
    print("Linus is", ages["Linus"])
```

## 4 · Adding, updating and deleting

Dictionaries are **mutable** (like lists): you can change them in place. Assigning to a key does one of two things:

- if the key **exists**, its value is replaced (update);
- if it **does not exist**, a new pair is added.

```python
stock = {"apples": 5, "pears": 2}

stock["plums"] = 12         # new key: added
stock["apples"] = 7         # existing key: replaced
stock["pears"] += 3         # read, add, store back
print(stock)
```

To delete an entry use `del`, or `pop`, which also hands the value back to you:

```python
del stock["plums"]
print(stock)

removed = stock.pop("pears")
print("removed", removed, "->", stock)

print(stock.pop("kiwis", 0))    # a second argument is the default if the key is missing
```

Two more handy tools: `update` copies in all the pairs of another dictionary (new keys are added, existing ones overwritten), and `clear` empties the dictionary.

```python
defaults = {"colour": "blue", "size": "M"}
choices = {"size": "L", "logo": True}

defaults.update(choices)
print(defaults)
```

## 5 · Missing keys: `get` and `setdefault`

Checking `in` before every lookup gets tedious. The method **`get`** looks up a key and, if it is missing, returns a **default** instead of raising an error. If you give no default, it returns `None`.

```python
ages = {"Ada": 36, "Grace": 45}

print(ages.get("Ada"))
print(ages.get("Alan"))             # None
print(ages.get("Alan", "unknown"))  # our own default
print(ages)                         # get never changes the dictionary
```

**`setdefault(key, default)`** is similar but also **stores** the default if the key was missing, and then returns whatever is stored. It shines when each value is a list that you want to append to (you will use it in the grouping examples below).

```python
settings = {"volume": 5}

print(settings.setdefault("volume", 10))   # already there: unchanged, returns 5
print(settings.setdefault("theme", "dark")) # missing: stored, returns "dark"
print(settings)
```

## 6 · Looping over a dictionary

A plain `for` loop over a dictionary gives you its **keys**. Three methods let you choose what you get:

| Expression | Gives you |
|---|---|
| `d` or `d.keys()` | the keys |
| `d.values()` | the values |
| `d.items()` | `(key, value)` pairs, which you can unpack in the `for` line |

```python
ages = {"Ada": 36, "Grace": 45, "Linus": 28}

for name in ages:
    print(name)
```

```python
for age in ages.values():
    print(age)

print(sum(ages.values()), max(ages.values()), min(ages.values()))
```

```python
for name, age in ages.items():      # unpacking, as in the tuples notebook
    print(f"{name} is {age}")
```

The results of `keys()`, `values()` and `items()` are special "view" objects; wrap them in `list(...)` if you want an actual list. To go through a dictionary in sorted key order, use `sorted`:

```python
print(list(ages.keys()))
print(list(ages.values()))
print(list(ages.items()))       # a list of (key, value) tuples

for name in sorted(ages):
    print(name, ages[name])
```

## 7 · Dictionaries as counters

Counting how often things occur is one of the most common dictionary jobs: the **key** is the thing, the **value** is its count. The pattern is "if I have seen it, add one; otherwise start at one". With `get` that is a single line:

```python
words = "the cat and the hat and the bat".split()

counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1

print(counts)
```

Read the key line as: "look up the current count (0 if new), add 1, store it back". The same works for letters in a string:

```python
letter_counts = {}
for letter in "banana":
    letter_counts[letter] = letter_counts.get(letter, 0) + 1
print(letter_counts)
```

To find the most frequent entry, combine this with the min/max-tracking pattern from Level 1:

```python
best_word = None
best_count = 0
for word, count in counts.items():
    if count > best_count:
        best_word = word
        best_count = count
print(best_word, best_count)
```

### Grouping with `setdefault`

A cousin of counting is **grouping**: the value is a *list* of everything that belongs under that key. `setdefault` creates the empty list the first time a key appears:

```python
names = ["Ada", "Alan", "Grace", "Barbara", "Gwen"]

by_initial = {}
for name in names:
    by_initial.setdefault(name[0], []).append(name)

print(by_initial)
```

`by_initial.setdefault(name[0], [])` returns the list stored under that initial (creating it if needed), and `.append(name)` adds to it. The same loop without `setdefault`, spelled out, in case it helps to see the steps:

```python
by_initial = {}
for name in names:
    initial = name[0]
    if initial not in by_initial:
        by_initial[initial] = []
    by_initial[initial].append(name)
print(by_initial)
```

## 8 · Nested dictionaries

A value can itself be a dictionary. That is how you describe a collection of records: an outer dictionary keyed by an id or name, and an inner dictionary for each record's fields.

```python
users = {
    "ada": {"name": "Ada Lovelace", "age": 36, "langs": ["Python", "C"]},
    "alan": {"name": "Alan Turing", "age": 41, "langs": ["Lisp"]},
}

print(users["ada"])
print(users["ada"]["name"])         # two lookups in a row
print(users["ada"]["langs"][0])     # dict -> dict -> list -> item
```

Update a nested value by following the same path, and loop with `items()` on the outer level:

```python
users["alan"]["age"] += 1
users["alan"]["langs"].append("Python")

for username, info in users.items():
    print(username, "->", info["name"], info["age"], info["langs"])
```

Careful: to add a *new* inner entry you have to create the inner dictionary first.

```python
users["grace"] = {"name": "Grace Hopper", "age": 45, "langs": []}
users["grace"]["langs"].append("COBOL")
print(users["grace"])
```

## 9 · Keys, copies and aliases

**Keys must be immutable.** Strings, numbers, booleans and tuples work as keys. Lists (and dictionaries) do not, because they can change. Python needs a key to stay the same so it can always find its entry again. (Notebook 16 returns to this when comparing collections.)

```python
grid = {(0, 0): "start", (2, 3): "treasure"}     # tuples as keys: fine
print(grid[(2, 3)])
```

**Aliases.** Dictionaries are mutable, so the lesson from lists applies: `b = a` does *not* copy, it gives the same dictionary a second name (an **alias**). Use `.copy()` for a separate (shallow) copy.

```python
a = {"x": 1}
b = a                   # alias: same dictionary
c = a.copy()            # new dictionary

b["x"] = 99
print(a)                # changed through b!
print(c)                # unaffected
print(a is b, a is c)
```

(`.copy()` is *shallow*: nested lists or dictionaries inside are still shared between the original and the copy. Be aware of it; the memory model is covered in depth much later in the course.)

## 10 · Common mistakes

### 1. Using a missing key with `[]`

`d[key]` raises `KeyError` when the key is absent. This often happens when counting and forgetting to start at zero: ⚠️ **Expected to fail.**

```python
counts = {}
for letter in "abca":
    counts[letter] += 1     # "a" has no count yet!
```

Fix: `counts[letter] = counts.get(letter, 0) + 1`, or check with `in` first.

### 2. Mixing up keys and values

Lookup goes from key to value, never the other way:

```python
ages = {"Ada": 36, "Grace": 45}
print(ages.get(36))     # 36 is a value, not a key -> None
```

### 3. Treating a dictionary like a list

There are no positions, so `ages[0]` looks for the *key* `0`: ⚠️ **Expected to fail.**

```python
print(ages[0])
```

### 4. Mutable keys

A list cannot be a key. Use a tuple instead: ⚠️ **Expected to fail.**

```python
bad = {[1, 2]: "pair"}
```

### 5. Changing a dictionary's size while looping over it

Adding or deleting keys during a `for` loop over the same dictionary breaks the loop: ⚠️ **Expected to fail.**

```python
stock = {"a": 0, "b": 5, "c": 0}
for item, qty in stock.items():
    if qty == 0:
        del stock[item]
```

Fix: loop over a **list copy** of the keys, `for item in list(stock):`, or build a new dictionary instead.

### 6. Expecting `b = a` to copy

See the alias example in section 9: use `a.copy()` when you need an independent dictionary.

## 12 · Recap and what's next

- A **dictionary** stores **key-value pairs**: `{"Ada": 36}`. You look values up by key, not by position. Keys are unique and must be immutable; order is the insertion order.
- `d[key]` looks up (a missing key is a `KeyError`); `d[key] = value` adds or updates; `del d[key]` and `d.pop(key)` remove.
- `key in d` tests keys. `d.get(key, default)` is a safe lookup; `d.setdefault(key, default)` looks up and stores a default if missing.
- Loop with `for key in d`, `d.values()` or `for key, value in d.items()`.
- **Counters** use `d[k] = d.get(k, 0) + 1`; **grouping** uses `d.setdefault(k, []).append(x)`.
- Values can be lists or other dictionaries, giving **nested** data. A dictionary is mutable, so `b = a` makes an alias; use `.copy()` for a separate dictionary.
