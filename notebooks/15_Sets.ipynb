## 1 · Why this matters

Lists remember order and happily hold duplicates. Dictionaries link keys to values. Sometimes you need neither: you only care about **which things are present**, each counted once.

- Which words appear in this text?
- Which friends do Ana and Ben have in common?
- Which of the numbers 1 to 100 did I forget to enter?

You *can* answer these with lists and loops, but there's a better tool. A **set** is a collection of **unique** items, modelled on the mathematical idea of a set, and it comes with built-in operations for exactly these questions. Sets are also very quick at answering "is this item in here?".

## 2 · Creating sets

Write a set with curly braces, or build one from any other collection with `set(...)`.

```python
fruits = {"apple", "banana", "cherry"}
print(len(fruits))
print(type(fruits))

numbers = set([3, 1, 2, 3, 3, 1])     # a list with repeats
print(sorted(numbers), len(numbers))   # each value is stored once

letters = set("mississippi")           # a string is a collection of characters
print(sorted(letters))
```

Three things to notice:

1. **No duplicates.** `3` and `1` appeared several times in the list, but the set keeps one of each.
2. **No meaningful order.** A set doesn't remember the order you gave. For strings, the order it prints can even differ from one run to the next. That's why the examples here use `sorted(...)` when they print: it gives you a list in a predictable order.
3. **No indexing or slicing.** There is no "item number 0" in a set.

### The empty set

`{}` is **not** an empty set. It's an empty *dictionary* (from the last notebook). For an empty set, use `set()`.

```python
empty_dict = {}
empty_set = set()
print(type(empty_dict))
print(type(empty_set))
```

### Which items can go in a set?

Items must be **immutable** (unchangeable), like numbers, strings and tuples. The same rule applies to dictionary keys, and for the same reason: Python needs to be sure the item won't change while it's stored. A list can change, so a list can't go into a set. (Notebook 16 gives the full picture.) Tuples are fine, which makes them handy for things like coordinates.

```python
visited = {(0, 0), (0, 1), (1, 1)}
print((0, 1) in visited)
print((5, 5) in visited)
```

## 3 · Membership, looping and size

The star operation is `in`, which for a set answers "is it present?" Everything else you know about collections still works: `len`, `for`, `sorted`, `min`, `max` and `sum`.

```python
primes = {2, 3, 5, 7, 11}

print(5 in primes)
print(9 in primes)
print(9 not in primes)
print(len(primes), max(primes), sum(primes))

for p in sorted(primes):      # sorted() gives a list, in a predictable order
    print("prime:", p)
```

Looping over a set directly (`for p in primes`) works too, but you can't rely on the order, so use `sorted(...)` when order matters.

## 4 · Adding and removing items

| Method | What it does |
|---|---|
| `s.add(x)` | add one item (does nothing if it's already there) |
| `s.update(other)` | add every item from another collection |
| `s.remove(x)` | remove `x`; **error** if it isn't there |
| `s.discard(x)` | remove `x` if present; **no error** if not |
| `s.pop()` | remove and return an **arbitrary** item |
| `s.clear()` | remove everything |

```python
guests = {"Ana", "Ben"}

guests.add("Cleo")
guests.add("Ana")                   # already there: nothing happens
print(sorted(guests))

guests.update(["Dev", "Eli", "Ben"])   # add several at once
print(sorted(guests))

guests.remove("Ben")
guests.discard("Zed")               # not present, and that's fine
print(sorted(guests))
```

Like list methods, these **change the set in place** and return `None`. Remember that sets are **mutable** (you can change them), even though their *items* must be immutable.

`pop()` on a set removes *some* item, and you don't get to choose which, so it isn't "the last one" as it is for lists. It's rarely what you want.

## 5 · Set operations

This is where sets shine. Take two groups of friends:

```python
ana = {"Ben", "Cleo", "Dev", "Eli"}
ben = {"Ana", "Cleo", "Eli", "Fay"}
```

### Union: everyone in either set (`|`)

Think "A **or** B". Each person still appears only once.

```python
print(sorted(ana | ben))
print(sorted(ana.union(ben)))     # the same thing, as a method
```

### Intersection: only those in both (`&`)

Think "A **and** B": the **common** friends.

```python
print(sorted(ana & ben))
print(sorted(ana.intersection(ben)))
```

### Difference: in the first but not the second (`-`)

Think "A **minus** B". **Order matters** here: `A - B` and `B - A` are different.

```python
print(sorted(ana - ben))     # Ana's friends that Ben doesn't know
print(sorted(ben - ana))     # Ben's friends that Ana doesn't know
```

### Symmetric difference: in exactly one of them (`^`)

Think "one or the other, but **not both**": everything except the overlap.

```python
print(sorted(ana ^ ben))
```

### Summary

With `A = {1, 2, 3, 4}` and `B = {3, 4, 5, 6}`:

| Operation | Operator | Method | Result |
|---|---|---|---|
| union | `A \| B` | `A.union(B)` | `{1, 2, 3, 4, 5, 6}` |
| intersection | `A & B` | `A.intersection(B)` | `{3, 4}` |
| difference | `A - B` | `A.difference(B)` | `{1, 2}` |
| symmetric difference | `A ^ B` | `A.symmetric_difference(B)` | `{1, 2, 5, 6}` |

A Venn diagram (two overlapping circles) is the best way to remember them. Draw one whenever you're unsure. The operators need two sets; the **methods** also accept any other collection, such as a list: `A.union([9, 10])`.

All four **return a new set** and leave `A` and `B` alone.

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A | B, A & B, A - B, A ^ B)
print(A)       # unchanged
```

### Comparing sets

You can also ask yes/no questions about how two sets relate:

```python
small = {1, 2}
big = {1, 2, 3, 4}

print(small <= big)                  # is small a subset of big?   (small.issubset(big))
print(big >= small)                  # is big a superset of small? (big.issuperset(small))
print(small.isdisjoint({7, 8}))      # no items in common?
print({1, 2} == {2, 1})              # equal: order doesn't matter
```

## 6 · Removing duplicates

The quickest way to remove repeated values from a list is to turn it into a set.

```python
raw = [4, 8, 4, 15, 8, 16, 23, 42, 15]
unique = set(raw)
print(len(raw), "items,", len(unique), "unique")
print(sorted(unique))
```

Because sets forget order, the result is no longer in the original order. If you want a list back, use `list(...)` or `sorted(...)`. If you need to keep the **original order** while dropping repeats, combine a list with a "have I seen this?" set:

```python
seen = set()
in_order = []
for n in raw:
    if n not in seen:
        seen.add(n)
        in_order.append(n)
print(in_order)
```

Another classic: **count the distinct words** in a sentence.

```python
sentence = "the cat and the dog and the bird"
words = sentence.split()
print(len(words), "words,", len(set(words)), "different words")
print(sorted(set(words)))
```

## 7 · Why is `in` fast on a set? (intuition only)

Looking for an item in a **list** means checking the items one by one, from the start, until you find it (or run out). A list with a million items might need a million checks.

A set works more like a **coat check** or the index at the back of a book. Python uses the item itself to work out *where* it should be stored, then goes straight to that spot instead of walking along the whole collection. The technique is called **hashing**, and it's also how dictionary keys are found quickly. You don't need the details now; just remember:

- `x in some_list` gets slower as the list grows;
- `x in some_set` stays quick however big the set is.

Here is a small experiment. (It uses the `time` module, which you haven't met yet, so treat that part as a preview and just read the results. Your numbers will differ.)

```python
import time

big_list = list(range(1_000_000))
big_set = set(big_list)

start = time.perf_counter()
for _ in range(100):
    999_999 in big_list
list_time = time.perf_counter() - start

start = time.perf_counter()
for _ in range(100):
    999_999 in big_set
set_time = time.perf_counter() - start

print(f"list: {list_time:.4f} seconds")
print(f"set:  {set_time:.6f} seconds")
```

**Rule of thumb:** if you do lots of "is it in there?" checks on a big collection and don't need the order or duplicates, use a set.

## 8 · Common mistakes

### 1. Expecting an order or an index

⚠️ **Expected to fail.**

```python
colours = {"red", "green", "blue"}
print(colours[0])
```

A set has no positions. Use `in` to test for presence, or `sorted(colours)` to get an ordered list (and then index *that*).

### 2. `{}` is a dictionary, not an empty set

⚠️ **Expected to fail.**

```python
tags = {}
tags.add("python")
```

`AttributeError`: a dict has no `add`. Write `tags = set()`.

### 3. `remove` on something that isn't there

⚠️ **Expected to fail.**

```python
team = {"Ana", "Ben"}
team.remove("Cleo")
```

`KeyError`. When you're not sure the item is present, use `discard`.

### 4. Putting a list in a set

⚠️ **Expected to fail.**

```python
groups = {[1, 2], [3, 4]}
```

`TypeError: unhashable type: 'list'`: items must be immutable. Use tuples: `{(1, 2), (3, 4)}`.

### 5. `+` doesn't work on sets

⚠️ **Expected to fail.**

```python
print({1, 2} + {3})
```

Use `|` (union) instead.

### 6. Forgetting that operations return a new set

```python
a = {1, 2}
b = {2, 3}
a | b                # computed, then thrown away
print(a)             # still {1, 2}
a = a | b            # keep the result...
print(sorted(a))
b.update({9})        # ...or change a set in place with update
print(sorted(b))
```

### 7. Surprise: sets hide duplicates you wanted to count
If you need to know **how many times** something appears, a set is the wrong tool: it throws the repeats away. Use a dictionary counter (notebook 14) instead.

## 10 · Recap and what's next

- A **set** holds **unique** items, with no order and no indexing. Create with `{1, 2, 3}` or `set(iterable)`; the empty set is `set()`, not `{}`.
- Items must be **immutable** (numbers, strings, tuples); a set itself is **mutable**.
- `in` tests membership, and is fast even on huge sets.
- `add`, `update`, `remove` (errors if missing), `discard` (doesn't), `pop`, `clear` change a set in place.
- **Union `|`, intersection `&`, difference `-`, symmetric difference `^`** return new sets. Also `<=`, `>=`, `isdisjoint` for comparing.
- `set(list)` removes duplicates (and the order). Use `sorted(...)` when you need a predictable order.
